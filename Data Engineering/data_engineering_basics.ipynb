# **📊 Data Engineering**

**The core idea:**

Think of building an ML model like cooking a meal in a restaurant.

- The chef (you, the ML engineer) needs fresh, clean, correctly portioned ingredients.
  
- But someone has to source the vegetables, wash them, chop them, and stock the kitchen before the chef even touches them.

That person is the **data engineer**.

> **Data engineering** is the work of getting data from where it is → to where it needs to be, in a form that's actually usable.

**A model trained on messy, incomplete, or badly structured data will always fail, no matter how clever the algorithm. Data engineering is what prevents that.**

---

## **The 3-step loop every data pipeline follows**

```python
[Source]  →  [Transform]  →  [Destination]
 Raw data      Clean it        Store it usably
```

This is sometimes called **ETL** — `Extract`, `Transform`, `Load`.

---

## **Types of Data**

1. **Structured Data**
   
Think of a **spreadsheet or database table** — rows, columns, every value has a defined place.

- Easy to query, filter, sort
- ML models love this — plug-and-play ready
- Stored in: SQL databases, CSV files, Excel

---

2. **Semi-Structured Data**

No rigid table, but **still has some organization** — like labels or tags (e.g. JSON, XML, YAML).

- Fields can vary between records
- Common when hitting APIs

---

3. **Unstructured Data**

**No predefined shape at all.**

- Text (emails, tweets, books)
- Images, audio, video
- Hardest to work with — you have to *create* structure before a model can use it

---

4. **Binary Data**
   
Raw computer-level files — **not human-readable.**

- Images stored as pixel bytes, audio/video files, model weights (`.pkl`, `.h5`)
- You don't read these directly — tools decode them for you

---

**Quick Mental Map**
```
Structured      → ready to use
Semi-structured → needs light processing
Unstructured    → needs heavy processing
Binary          → needs decoding
```

## **Data Engineering Concepts**

## **1. Data Mining**
Digging through large raw data to find **hidden patterns** — finds the questions *and* answers at the same time.

| Regular ML | Data Mining |
|---|---|
| Clear question, predict | Exploring, no fixed question |
| Supervised | Often unsupervised |

**Techniques:** Clustering · Association rules · Anomaly detection · Classification

---

## **2. Big Data**
When data becomes too large for one machine — like a tap vs a river.

**The 3 Vs:**
```
Volume   → how MUCH (terabytes, petabytes)
Velocity → how FAST (real-time, millions of rows/sec)
Variety  → how MANY types (structured + unstructured + binary)
```

**Solution:** Split data across many machines, process in parallel, combine results.
Tools: Apache Spark · AWS S3 · Google Cloud

### **🛩️ Fighter Jet Example**
Hundreds of sensors firing every millisecond = Big Data (Velocity).
But a jet has no cluster or internet — so AI works in **two phases:**

```
GROUND → Train on Big Data using clusters
IN-AIR → Run compressed pre-trained model (Edge AI)

Training = heavy (ground)
Inference = light (jet)
```

> Your phone's Face ID works the same way — no internet, runs on-chip.

---

## **3. Data Pipelines**
A system that moves raw data through stages until it's usable — like an oil pipeline.

```
[Source] → [Ingest] → [Process] → [Store] → [Serve]
```

```
Without pipeline → manually clean data every time 😓
With pipeline    → data flows and updates automatically ✅
```

**Three types:**

| Type | When | Example |
|---|---|---|
| Batch | Scheduled | Process sales every morning |
| Streaming | Real-time | Fraud detection per transaction |
| Lambda | Both | Historical + live data |

```
Data Mining    → find patterns in data
Big Data       → handle data too large for one machine
Data Pipeline  → move and prepare the data automatically
```

## **Types of Databases**

---

## **1. Relational Databases (SQL)**
Data lives in tables, rows and columns. Tables are linked through relationships.

```
CUSTOMERS table          ORDERS table
─────────────────        ──────────────────────
ID  | Name               ID  | CustomerID | Item
1   | Eakempreet         1   | 1          | Pizza
2   | Rahul              2   | 1          | Burger
```

- Examples: **MySQL, PostgreSQL, SQLite**
- Best for: structured, consistent data
- Language: **SQL**

---

## **2. Non-Relational Databases (NoSQL)**
No strict tables. Data stored flexibly as JSON documents.

```json
{
  "name": "Eakempreet",
  "age": 22,
  "orders": ["Pizza", "Burger", "Fries"]
}
```

One user can have 3 orders, another 10 — no problem.

- Examples: **MongoDB, Firebase, CouchDB**
- Best for: flexible, fast-changing data
- Used in: apps, social media, real-time features

### **SQL vs NoSQL**
```
SQL                           NoSQL
─────────────────             ──────────────────────
Strict structure              Flexible structure
Tables + rows                 Documents / JSON
Data rarely changes           Data changes often
Bank records                  Social media posts
```

---

## **3. Columnar Databases**
Stores data **column by column** instead of row by row.

```
ROW STORAGE                      COLUMNAR STORAGE
────────────────────────         ────────────────────────
[Eakempreet | 22 | Delhi]        [Eakempreet, Rahul]  ← Names
[Rahul      | 25 | Mumbai]       [22, 25]             ← Ages
                                 [Delhi, Mumbai]      ← Cities
```

When you only need ages — columnar reads just that one block. Massively faster for analysis.

> Each value keeps its **index** across all columns — so the database stitches the right values together when needed.
> `Index 0 → Eakempreet | 22 | Delhi`

### **The School Register Analogy**
```
ROW WAY       → one page per student, all their info on that page
COLUMNAR WAY  → one page for all names, one page for all ages
```

- Examples: **Amazon Redshift, Google BigQuery**
- Best for: analytics, ML datasets, Big Data analysis

---

## **4. NewSQL Databases**
Best of both worlds — SQL structure + NoSQL scale.

```
SQL                 NoSQL               NewSQL
───────────         ──────────          ──────────────
Structured    +     Scales across   =   Structured AND
Reliable            many machines       scales massively
```

- Examples: **Google Spanner, CockroachDB**
- Best for: large scale apps that still need strict structure
- Real world: Google uses Spanner to run Gmail and Google Photos globally

---

## **All 4 Together**

| Type | Best For | Examples |
|---|---|---|
| **Relational (SQL)** | Structured, consistent data | MySQL, PostgreSQL |
| **Non-Relational (NoSQL)** | Flexible, fast-changing data | MongoDB, Firebase |
| **Columnar** | Analytics, ML, Big Data | Redshift, BigQuery |
| **NewSQL** | Large scale + structure | Spanner, CockroachDB |

## **Where These Fit**

> Like cars — Sedan/SUV are general purpose. F1 car/Ambulance are built for one job.

---

## **1. Search Databases**
Built for finding **relevant** matches, not just exact ones.

```
Normal DB  → "find name = Eakempreet"       ← exact match
Search DB  → "find anything about cricket"   ← relevant matches
```

- Examples: **Elasticsearch, Apache Solr**
- Best for: search bars, log analysis, recommendations
- Used in: Amazon product search, Spotify song search

---

## **2. Computation Databases (In-Memory)**
Loads everything into **RAM** instead of reading from disk.

```
Disk database  → data on hard drive → fetch when needed → slow
In-Memory DB   → data lives in RAM  → always ready     → fast ⚡
```

Tradeoff — RAM is expensive and data is lost if power goes off.

- Examples: **Redis, Memcached**
- Best for: real-time leaderboards, session data, caching
- Used in: gaming scores, live cricket scoreboards

---

## **OLTP vs OLAP**
Not new database types — two **modes of using** a database.

---

## **OLTP — Online Transaction Processing**
Built for **fast, small, frequent operations.**

Zomato order being placed right now:
- Insert new order ✅
- Update payment status ✅
- Delete cancelled order ✅

```
OLTP
→ Many users
→ Small operations
→ Real-time
→ Write heavy
→ Examples: MySQL, PostgreSQL
```

---

## **OLAP — Online Analytical Processing**
Built for **slow, large, complex analysis.**

Zomato data team asking:
- "Which city had most orders last quarter?"
- "What is the average delivery time by restaurant?"

```
OLAP
→ Few users (analysts, ML engineers)
→ Huge operations across millions of rows
→ Not real-time
→ Read heavy
→ Examples: Redshift, BigQuery
```

---

## **Side by Side**

```
OLTP                        OLAP
────────────────────        ────────────────────
Bank transaction            Bank monthly report
Placing an order            Analysing all orders
Insert / Update / Delete    Select / Aggregate
Milliseconds                Seconds to minutes
MySQL, PostgreSQL            Redshift, BigQuery
```

## **How It All Connects**

```
OLTP captures the live data
         ↓
Data pipeline moves it
         ↓
OLAP analyses it
         ↓
ML model trains on it
```

> OLTP is the **kitchen** taking orders.
> OLAP is the **manager** reviewing the day's performance.

## **Hadoop, HDFS and MapReduce**

---

## **Hadoop vs HDFS — Not the Same**

Hadoop is the **entire system.** HDFS is just one part of it.

```
HADOOP (the full system)
├── HDFS          → stores the data
├── MapReduce     → processes the data
└── YARN          → manages resources (who gets what machine)
```

> Hadoop is the **car.** HDFS is just the **fuel tank.**

---

## **HDFS — Hadoop Distributed File System**

Solves two problems — **safety** and **speed.**

### **The Egg Analogy**
12 eggs in 1 basket → basket drops → all lost ❌

HDFS says — split across 4 baskets:
```
Basket 1 → 3 eggs
Basket 2 → 3 eggs
Basket 3 → 3 eggs
Basket 4 → 3 eggs
```
One basket drops → only 3 eggs lost. The rest are safe ✅

**Baskets = machines | Eggs = chunks of your data**

---

### **Chunks — Solving Speed**
Your data is split into pieces called **chunks** and spread across many machines.

```
WITHOUT chunks
──────────────
1 machine processes 100GB → takes 10 hours

WITH chunks across 1000 machines
──────────────────────────────────
Each machine processes 0.1GB simultaneously → takes minutes ⚡
```

---

### **3 Copies — Solving Safety**
Each chunk is copied 3 times and stored on 3 different machines.

```
12 chunks of original data → stored with 3 copies each

Chunk 1 → Machine 1, Machine 3, Machine 5
Chunk 2 → Machine 2, Machine 4, Machine 6
...and so on
```

Machine 1 dies → Chunk 1 still safe on Machine 3 and 5 ✅

### **The Key Point**
```
1000 machines → 3000 chunks stored
BUT only 1000 unique chunks of real data
The other 2000 are safety backups
sitting idle until a machine dies
```

Yes — 100GB becomes 300GB stored.
That is the **deliberate tradeoff:**
```
Pay 3x storage cost
in exchange for
zero data loss ✅
```
Storage is cheap. Losing data is catastrophic.

---

### **Two Problems, Two Solutions**
```
PROBLEM                    SOLUTION
────────────────────────────────────────
Machine dies, data lost  → 3 copies (safety)
Data too big to process  → chunks across machines (speed)
```

---

## **MapReduce — The Processing Layer**

Once data is stored across machines — MapReduce processes it in two steps.

### **MAP — process locally**
Each machine processes only its own chunk simultaneously.
```
Machine 1 → processes its chunk → local result
Machine 2 → processes its chunk → local result
Machine 3 → processes its chunk → local result
```

### **REDUCE — combine results**
All local results merged into one final answer.
```
Local results → [REDUCER] → Final Answer
```

---

### **Real World Example**
Count orders from Delhi across 1 billion rows:
```
MAP phase
─────────
Machine 1 → finds 2M Delhi orders
Machine 2 → finds 3M Delhi orders
Machine 3 → finds 1M Delhi orders

REDUCE phase
────────────
2M + 3M + 1M = 6M total Delhi orders ✅
```

---

## **HDFS + MapReduce Together**

```
HDFS          →    stores split data across machines
MapReduce     →    processes in parallel, combines results
```

> HDFS is the **warehouse** with shelves across many rooms.
> MapReduce is the **team of workers** each searching their own room, then reporting back.

## **Apache Spark and Apache Flink**

---

## **Apache Spark**

Does the same job as MapReduce — processes Big Data across machines.
But keeps everything in **RAM instead of disk — up to 100x faster.**

### **Why Faster**
```
MapReduce
─────────
Step 1 → process → save to disk
Step 2 → read from disk → process → save to disk
😓 slow

Spark
─────
Step 1 → process → keep in RAM
Step 2 → process → keep in RAM
Step 3 → final result
⚡ fast
```

### **Speed Difference**
```
MapReduce → 10 hours
Spark     → 10 minutes (up to 100x faster)
```

### **What Spark is Used For**
- Processing massive datasets
- Real-time streaming data
- ML on Big Data via **MLlib** (Spark's own ML library)
- SQL queries on Big Data

> MapReduce = **old slow train**
> Spark = **bullet train on the same tracks**

Used by: Netflix, Uber, Airbnb

---

## **Apache Flink**

Spark processes data in **micro batches** — collects a few seconds of data, then processes.
Flink processes **each event the instant it arrives.** True real-time.

### **The Cashier Analogy**
```
Spark  → serves customers in groups of 10
         "wait for 10 people then process"

Flink  → serves each customer the exact moment they arrive
         zero waiting ⚡
```

### **Real World Example — Fraud Detection**
```
Spark  → collects 5 seconds of transactions
         then checks for fraud
         → fraud already happened 😓

Flink  → transaction arrives
         checked instantly
         fraud blocked before it completes ✅
```

---

## **Spark vs Flink**

```
Spark                        Flink
─────────────────────        ─────────────────────
Near real-time               True real-time
Micro batches                Event by event
Great for ML                 Great for live systems
Netflix recommendations      Bank fraud detection
```

---

## **All Three Together**

```
MapReduce  → original, slow, writes to disk
Spark      → faster, keeps in RAM, micro batches
Flink      → fastest, true real-time, event by event
```

> All three do the same job — processing Big Data.
> Just at different speeds for different needs.

## **Stream Processing and Apache Kafka**

---

## **Stream Processing**

Two ways to process data:

```
BATCH processing    → collect data, wait, then process
STREAM processing   → process data the instant it arrives
```

### **The Laundry Analogy**
```
BATCH    → wash clothes once a week
           collect all dirty clothes → wash together

STREAM   → wash each piece of clothing
           the moment it gets dirty
```

### **When You Need Streaming**
When a delay of even seconds causes real damage:

```
Fraud detection      → 5 second delay = fraud succeeded
Crash detection      → 5 second delay = car already crashed
Stock trading        → 5 second delay = money lost
Live recommendations → delay = user already left the page
```

---

## **Apache Kafka**

Kafka is a **message passing system** — sits in the middle between data producers and consumers.

### **The Restaurant Analogy**
```
WITHOUT Kafka
─────────────
Customers shout orders directly at chefs
10 customers at once → chaos, orders lost ❌

WITH Kafka
──────────
Customers → write on ticket board → chefs read from board
Organised, nothing lost, chefs work at their own pace ✅
```

> **Kafka is the ticket board.**

---

### **Three Core Concepts**

```
PRODUCER    → anything that sends data
              (app, sensor, website click, transaction)

TOPIC       → a named channel data flows through
              (like a queue — "orders", "payments", "clicks")

CONSUMER    → anything that reads data
              (ML model, fraud detector, dashboard)
```

### **How It Flows**
```
[Producer]          [Kafka]             [Consumer]
──────────          ───────             ──────────
User clicks    →    Topic:clicks    →   Recommendation engine
Bank transfer  →    Topic:payments  →   Fraud detector
Sensor fires   →    Topic:sensors   →   ML model
```

---

### **Why Kafka**

```
Fast        → handles millions of events per second
Reliable    → messages saved to disk, never lost
Scalable    → add more machines as traffic grows
Replayable  → consumers can re-read old messages
```

### **Real World — Uber**
```
Drivers  → send location every second  → Kafka → map updates
Riders   → request rides               → Kafka → matching system
Payments → process on trip end         → Kafka → fraud detection
```
All happening simultaneously through Kafka topics.

---

## **Kafka in the Full Pipeline**

```
[Live Events]
      ↓
   KAFKA
(catches everything)
      ↓
  SPARK / FLINK
(processes the stream)
      ↓
  S3 / REDSHIFT
(stores clean data)
      ↓
  ML MODEL
(trains or predicts)
```
