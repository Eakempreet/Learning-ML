What does the transpose do?

````python
# Create two arrays of random integers between 0 to 10
# one of size (3, 3) the other of size (3, 2)

rand1 = np.random.randint(0, 10, size=(3, 3))
rand2 = np.random.randint(0, 10, size=(3,2))

rand1, rand1.shape, rand2, rand2.shape
````

````python
# Perform a dot product on the two newest arrays you created
rand1 @ rand2
````

````python
# Create two arrays of random integers between 0 to 10
# both of size (4, 3)
np.random.seed(42)

random1 = np.random.randint(0, 10, size=(4, 3))
random2 = np.random.randint(0, 10, size=(4, 3))

random1, random1.shape, random2, random2.shape
````

````python
# Perform a dot product on the two newest arrays you 

#random1 @ random2

# ValueError: matmul: Input operand 1 has a mismatch in its core dimension 0, with gufunc signature (n?,k),(k,m?)->(n?,m?) (size 4 is different from 3)
````

It doesn't work. How would you fix it?

````python
# Take the latest two arrays, perform a transpose on one of them and then perform 
# a dot product on them both

random1 @ random2.T
````

````python
np.dot(random1, random2.T)
````

Notice how performing a transpose allows the dot product to happen.

Why is this?
