### ***Reshaping and Transposing***

#### **What is Reshape?**
Reshape changes the shape of an array. The data stays the same, but the layout changes. For example, change 6 numbers in a line to 2 rows and 3 columns.

**Why use Reshape?**
- Make arrays same size to do math together
- Get data in right format for models
- Rearrange without losing data

#### **What is Transpose?**
Transpose flips an array. Rows become columns, columns become rows. Use `.T` to do it.

**Why use Transpose?**
- Flip data from rows to columns
- Get data ready for calculations
- Fix array shape when needed

## **Element-Wise Multiplication vs Dot Product**

### **Example: Two Small Matrices**

```
mat1 = [1  2]      mat2 = [3  4]
       [5  6]             [7  8]
```

---

### **1️⃣ ELEMENT-WISE MULTIPLICATION (Hadamard Product)**
**Use:** `mat1 * mat2`

Multiply matching positions, one by one:

```
[1  2]   *   [3  4]   =   [1×3  2×4]   =   [3   8]
[5  6]       [7  8]       [5×7  6×8]       [35  48]
```

**Result shape:** Same as input `(2, 2)`

---

### **2️⃣ DOT PRODUCT (Matrix Multiplication)**
**Use:** `mat1 @ mat2` or `np.dot(mat1, mat2)`

**RULE:** `(rows1, cols1) @ (rows2, cols2)` → `(rows1, cols2)`

**Constraint:** `cols1 must equal rows2`

For dot product: Multiply rows of mat1 by columns of mat2, then sum:

```
mat1 shape = (2, 2)
mat2 shape = (2, 2)  ✅ Can multiply! (2 == 2)

Result = [1×3 + 2×7,    1×4 + 2×8]   =   [17  20]
         [5×3 + 6×7,    5×4 + 6×8]       [57  68]
```

**Result shape:** `(2, 2)` (rows of mat1, cols of mat2)

---

### **Key Difference**

| Operation | Symbol | Formula | Result |
|-----------|--------|---------|--------|
| **Element-Wise** | `*` | Position by position | Same shape |
| **Dot Product** | `@` | Rows × Columns (sum) | (rows1, cols2) |

**Element-wise:** `[3, 8, 35, 48]` (just multiply)  
**Dot product:** `[17, 20, 57, 68]` (multiply + sum)

````python
# Since dim1 of mat1 == dim0 of ma2

mat_trans = np.transpose(mat2)
````

````python
# Dot prdoct

np.dot(mat1, mat_trans)
````
