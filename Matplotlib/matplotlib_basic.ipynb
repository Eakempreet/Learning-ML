# Matplotlib: pyplot vs OO Interface 🎨

---

## 🎯 **Two Ways to Make Plots**

### **1. pyplot = Quick & Simple**
````python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [1, 4, 9])
plt.title("My Plot")
plt.show()
````

**Think:** Giving commands to a magic canvas
- "Draw a line!"
- "Add a title!"

---

### **2. OO = Powerful & Clear**
````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot([1, 2, 3], [1, 4, 9])
ax.set_title("My Plot")
plt.show()
````

**Think:** You own the canvas, you control it
- "My canvas (`ax`), draw this line"
- "My canvas, add this title"

---

## 📊 **Quick Comparison**

| Feature | pyplot | OO |
|---------|--------|-----|
| **Easy to learn?** | ✅ Yes | ⚠️ Slightly harder |
| **Good for 1 plot?** | ✅ Yes | ✅ Yes |
| **Good for many plots?** | ❌ Confusing | ✅ Perfect |
| **Professional code?** | ⚠️ OK | ✅ Best |

---

## 💡 **Remember This**

### **pyplot = Training Wheels** 🚲
- Start here!
- Great for learning
- One simple plot at a time

### **OO = Real Bike** 🏍️
- Use this later!
- Professional code
- Multiple plots, complex figures

`````python
# Creating a plot with the OO verison, confusing way first
# 1st method

fig = plt.figure()  # Creates a figure
ax = fig.add_subplot() # add some axes
plt.show()
`````

- `fig` = the whole page/window (canvas)
- `ax` = the graph box drawn inside that page
- `plt.show()` = “now show me the page”

```text
Figure (fig)
┌──────────────────────────────┐
│                              │
│   Axes/Subplot (ax)          │
│   ┌──────────────────────┐   │
│   │                      │   │
│   │   (where data goes)  │   │
│   │                      │   │
│   └──────────────────────┘   │
│                              │
└──────────────────────────────┘
```

`````python
# 2nd method

fig = plt.figure()
ax = fig.add_axes([1, 1, 1, 1]) # type: ignore
ax.plot(x, y)
plt.show()
`````

- `fig = plt.figure()` → creates the full canvas (whole page)
- `add_axes([left, bottom, width, height])` uses figure coordinates from 0 to 1.
So [1, 1, 1, 1] means:
  - left starts at far right edge,
  - bottom starts at top edge,
  - width and height are full size.
  - This places axes mostly outside the visible figure area.


```text
(0,1) ┌──────────────────────────┐ (1,1)
      │                          │
      │        visible area      │
      │                          │
(0,0) └──────────────────────────┘ (1,0)

add_axes([1,1,1,1]) starts here ↑ (top-right corner)
so most of that axes goes out of bounds.
```

`````python
# Easier and more robust going forward (what we're going to use)

fig, ax = plt.subplots()
ax.plot(x, y); # add some data
type(fig), type(ax)
`````

- `fig` = whole canvas/window (the paper).
- `ax` = one plot area inside that canvas (the graph box).
- `ax.plot(x, y)` = draw your data on that specific plot area.

```text
Figure (fig)
┌──────────────────────────────┐
│      Axes (ax)               │
│   ┌──────────────────────┐   │
│   │   line from x,y      │   │
│   └──────────────────────┘   │
└──────────────────────────────┘
```

`````python
fig, ax = plt.subplots()
ax.plot(x, [50, 100, 200, 250]);
`````

# **Matplotlib example workflow**

`````python
# 0. Import matplotlib and get it ready for plotting in 
%matplotlib inline
import matplotlib.pyplot as plt

# 1. Prepare the data
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10, 5))  # figsize=(width, height)

# 3. Plot the data
ax.plot(x, y);

# 4. Customize the plot
ax.set(title="Simple plot",
       xlabel="x-axis",
       ylabel="y-axis")

# 5. Save and show (you save the whole figure)
fig.savefig(r"../assets/images/sample-plot.png")
`````

`````python
# Example 2

# 0. import matplotlib
%matplotlib inline
import matplotlib.pyplot as plt

# 1. Prepare data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 2. Setup plot
fig, ax = plt.subplots()

# 3. Plot the data
ax.plot(x, y)

# 4. Customize
ax.set_title("Sin Wave")
ax.set_xlabel("Numbers")
ax.set_ylabel("Sin")          # Previous example method is also good

# 5. Save and show
fig.savefig(r"../assets/images/Exercise_Sin_Wave")
`````

## **Making figures with NumPy arrays**

**We Want:**
- Line plot - (default one)
- Scatter plot
- Bar plot
- Histogram 
- Subplots

`````python
import numpy as np
`````

`````python
# Creating some data

x = np.linspace(0, 10, 100)
x[:10]
`````

`````python
# Plot the Data and create a line plot
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(x, x**2);
`````

`````python
# Use same data to make Scatter plot

fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
`````

`````python
# Another Scatter plot
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
`````

`````python
# Make a plot from dictionary
nut_butter_prices = {"Almond butter" : 10,
                     "Peanut butter" : 8,
                     "Cashew butter" : 12}
fig, ax = plt.subplots()
ax.scatter(nut_butter_prices.keys(), nut_butter_prices.values()) # type: ignore
`````

`````python
# Make a plot from dictionary (Histogram)
nut_butter_prices = {"Almond butter" : 10,
                     "Peanut butter" : 8,
                     "Cashew butter" : 12}
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())   # type: ignore # (X,Y)
ax.set(title="Butter Wala Store", ylabel="Prices($)");
`````
