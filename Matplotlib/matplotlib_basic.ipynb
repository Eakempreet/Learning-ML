# Matplotlib: pyplot vs OO Interface 🎨

---

## 🎯 **Two Ways to Make Plots**

### **1. pyplot = Quick & Simple**
````python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [1, 4, 9])
plt.title("My Plot")
plt.show()
````

**Think:** Giving commands to a magic canvas
- "Draw a line!"
- "Add a title!"

---

### **2. OO = Powerful & Clear**
````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot([1, 2, 3], [1, 4, 9])
ax.set_title("My Plot")
plt.show()
````

**Think:** You own the canvas, you control it
- "My canvas (`ax`), draw this line"
- "My canvas, add this title"

---

## 📊 **Quick Comparison**

| Feature | pyplot | OO |
|---------|--------|-----|
| **Easy to learn?** | ✅ Yes | ⚠️ Slightly harder |
| **Good for 1 plot?** | ✅ Yes | ✅ Yes |
| **Good for many plots?** | ❌ Confusing | ✅ Perfect |
| **Professional code?** | ⚠️ OK | ✅ Best |

---

## 💡 **Remember This**

### **pyplot = Training Wheels** 🚲
- Start here!
- Great for learning
- One simple plot at a time

### **OO = Real Bike** 🏍️
- Use this later!
- Professional code
- Multiple plots, complex figures

---




`````python
# Creating a plot with the OO verison, confusing way first
# 1st method

fig = plt.figure()  # Creates a figure
ax = fig.add_subplot() # add some axes
plt.show()
`````

- `fig` = the whole page/window (canvas)
- `ax` = the graph box drawn inside that page
- `plt.show()` = “now show me the page”

```text
Figure (fig)
┌──────────────────────────────┐
│                              │
│   Axes/Subplot (ax)          │
│   ┌──────────────────────┐   │
│   │                      │   │
│   │   (where data goes)  │   │
│   │                      │   │
│   └──────────────────────┘   │
│                              │
└──────────────────────────────┘
```

`````python
# 2nd method

fig = plt.figure()
ax = fig.add_axes([1, 1, 1, 1]) # type: ignore
ax.plot(x, y)
plt.show()
`````

- `fig = plt.figure()` → creates the full canvas (whole page)
- `add_axes([left, bottom, width, height])` uses figure coordinates from 0 to 1.
So [1, 1, 1, 1] means:
  - left starts at far right edge,
  - bottom starts at top edge,
  - width and height are full size.
  - This places axes mostly outside the visible figure area.


```text
(0,1) ┌──────────────────────────┐ (1,1)
      │                          │
      │        visible area      │
      │                          │
(0,0) └──────────────────────────┘ (1,0)

add_axes([1,1,1,1]) starts here ↑ (top-right corner)
so most of that axes goes out of bounds.
```

`````python
# Easier and more robust going forward (what we're going to use)

fig, ax = plt.subplots()
ax.plot(x, y); # add some data
type(fig), type(ax)
`````

- `fig` = whole canvas/window (the paper).
- `ax` = one plot area inside that canvas (the graph box).
- `ax.plot(x, y)` = draw your data on that specific plot area.

```text
Figure (fig)
┌──────────────────────────────┐
│      Axes (ax)               │
│   ┌──────────────────────┐   │
│   │   line from x,y      │   │
│   └──────────────────────┘   │
└──────────────────────────────┘
```

`````python
fig, ax = plt.subplots()
ax.plot(x, [50, 100, 200, 250]);
`````

# **Matplotlib example workflow**

`````python
# 0. Import matplotlib and get it ready for plotting in 
%matplotlib inline
import matplotlib.pyplot as plt

# 1. Prepare the data
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10, 5))  # figsize=(width, height)

# 3. Plot the data
ax.plot(x, y);

# 4. Customize the plot
ax.set(title="Simple plot",
       xlabel="x-axis",
       ylabel="y-axis")

# 5. Save and show (you save the whole figure)
fig.savefig(r"../assets/images/sample-plot.png")
`````

`````python
# Example 2

# 0. import matplotlib
%matplotlib inline
import matplotlib.pyplot as plt

# 1. Prepare data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 2. Setup plot
fig, ax = plt.subplots()

# 3. Plot the data
ax.plot(x, y)

# 4. Customize
ax.set_title("Sin Wave")
ax.set_xlabel("Numbers")
ax.set_ylabel("Sin")          # Previous example method is also good

# 5. Save and show
fig.savefig(r"../assets/images/Exercise_Sin_Wave")
`````

## **Subplots**
###  **Two options for Subplots**

`````python
# Subplot option 1

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2,
                                             ncols=2,
                                             figsize=(10,5))

# Plot to each axis
ax1.plot(x, x/2)
ax2.scatter(np.random.random(100), np.random.random(100))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000));
`````

`````python
# Subplot option 2
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize=(10,5))
# Plot to each different index
ax[0, 0].plot(x, x/2)
ax[0, 1].scatter(np.random.random(50), np.random.random(50))
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1, 1].hist(np.random.randn(1000));
`````

`````python
np.random.seed(123456)

ts = pd.Series(np.random.randn(1000), index=pd.date_range("1/1/2026", periods=1000))
print(ts)
ts.plot();
`````

`ts = ts.cumsum()`
- **What it does:** Calculate cumulative sum
```text
Original:        Cumulative Sum:
1                1
2                1+2 = 3
3                1+2+3 = 6
4                1+2+3+4 = 10
```

**Why use it?**

- Shows trend over time 📈
- Converts random walk into a smooth path
- Useful for finance/stock price simulation

`````python
ts = ts.cumsum()
ts.plot()
`````
