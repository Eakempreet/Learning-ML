### ***Reshaping and Transposing***

#### **What is Reshape?**
Reshape changes the shape of an array. The data stays the same, but the layout changes. For example, change 6 numbers in a line to 2 rows and 3 columns.

**Why use Reshape?**
- Make arrays same size to do math together
- Get data in right format for models
- Rearrange without losing data

#### **What is Transpose?**
Transpose flips an array. Rows become columns, columns become rows. Use `.T` to do it.

**Why use Transpose?**
- Flip data from rows to columns
- Get data ready for calculations
- Fix array shape when needed

## **Element-Wise Multiplication vs Dot Product**

### **Example: Two Small Matrices**

```
mat1 = [1  2]      mat2 = [3  4]
       [5  6]             [7  8]
```

---

### **1️⃣ ELEMENT-WISE MULTIPLICATION (Hadamard Product)**
**Use:** `mat1 * mat2`

Multiply matching positions, one by one:

```
[1  2]   *   [3  4]   =   [1×3  2×4]   =   [3   8]
[5  6]       [7  8]       [5×7  6×8]       [35  48]
```

**Result shape:** Same as input `(2, 2)`

---

### **2️⃣ DOT PRODUCT (Matrix Multiplication)**
**Use:** `mat1 @ mat2` or `np.dot(mat1, mat2)`

**RULE:** `(rows1, cols1) @ (rows2, cols2)` → `(rows1, cols2)`

**Constraint:** `cols1 must equal rows2`

For dot product: Multiply rows of mat1 by columns of mat2, then sum:

```
mat1 shape = (2, 2)
mat2 shape = (2, 2)  ✅ Can multiply! (2 == 2)

Result = [1×3 + 2×7,    1×4 + 2×8]   =   [17  20]
         [5×3 + 6×7,    5×4 + 6×8]       [57  68]
```

**Result shape:** `(2, 2)` (rows of mat1, cols of mat2)

---

### **Key Difference**

| Operation | Symbol | Formula | Result |
|-----------|--------|---------|--------|
| **Element-Wise** | `*` | Position by position | Same shape |
| **Dot Product** | `@` | Rows × Columns (sum) | (rows1, cols2) |

**Element-wise:** `[3, 8, 35, 48]` (just multiply)  
**Dot product:** `[17, 20, 57, 68]` (multiply + sum)

````python
# Since dim1 of mat1 == dim0 of ma2

mat_trans = np.transpose(mat2)  
````

````python
# Dot prdoct

np.dot(mat1, mat_trans)
````

````python
mat1@mat_trans
````

````python
mat3 = np.dot(mat1, mat2.T)
mat3
````

### **Why and When Use Dot Product?**

**WHY:**
- Get **one final number** from many numbers (like calculating a score)
- Core operation in **all ML models** — every prediction uses dot product
- Combine features with weights: `prediction = data @ weights`

**WHEN:**
- Making predictions (house price, test score, any ML model output)
- Measuring similarity between two vectors
- Neural networks (every layer uses dot product)

**Example:**  
House features: `[1500 sqft, 3 bedrooms, 10 years old]`  
Model weights: `[200, 5000, -1000]`  
Price = `1500×200 + 3×5000 + 10×(-1000)` = **$305,000** ← That's dot product!

## **IMPORTANT: 1D Arrays Don't Change Shape in Dot Product**

**Common Confusion:**
- Does `(3,)` convert to `(1, 3)` when doing dot product? **NO!**

**What Actually Happens:**

```
1D Array stays 1D:
prices.shape = (3,)
result = prices @ sales_amount.T
result.shape = (5,)  ← Still 1D! No conversion!

2D Array stays 2D:
prices_2d.shape = (1, 3)
result_2d = prices_2d @ sales_amount.T
result_2d.shape = (1, 5)  ← Still 2D!
```

**The Rule:**
- NumPy **does NOT convert shapes**
- It just **handles the math correctly**
- 1D stays 1D, 2D stays 2D

**In our example:**
```
prices = [10, 8, 12]          shape (3,)
prices @ sales_amount.T       
= [revenue_mon, revenue_tue, revenue_wed, revenue_thu, revenue_fri]
shape (5,) ← 1D output!
```

NumPy is smart about dimensions. No hidden conversions! ✅

## **ROWS vs COLUMNS Confusion**

**Visual Example:**
```
Matrix = [[1,  2,  3],
          [4,  5,  6]]

ROWS (horizontal):
  Row 1 = [1, 2, 3]
  Row 2 = [4, 5, 6]

COLUMNS (vertical):
  Col 1 = [1, 4]
  Col 2 = [2, 5]
  Col 3 = [3, 6]

Shape (2, 3) means:
  2 rows
  3 columns
```

---

### **How Rows & Columns Matter in Dot Product**

**Rule:** `(rows1, cols1) @ (rows2, cols2)` ← **cols1 MUST equal rows2**

**Example:**
```
Matrix A: shape (2, 3)  ← 2 rows, 3 columns
Matrix B: shape (3, 5)  ← 3 rows, 5 columns

A @ B works because 3 (cols of A) == 3 (rows of B) ✅

Result: (2, 5)  ← Keep rows from A, keep cols from B
```

---

### **In Our Butter Exercise**

```
prices.shape = (3,)                    ← 3 prices
sales_amount = (5, 3)                  ← 5 days, 3 types per day

We transpose: sales_amount.T = (3, 5)  ← Now 3 rows, 5 columns

prices @ sales_amount.T
(3,) @ (3, 5)
       ↓
   Match! ✅
   
Result: (5,)  ← 5 daily revenues
```

**Why transpose?** We rearranged the data so:
- 3 rows matched 3 prices
- 5 columns gave us 5 days

### **Axis**

- Axis tells NumPy which direction to perform an operation (like sum, mean, max, etc.)
- **Think of it as:** "Which way should I collapse/combine the data?"


> axis=0  →  Go DOWN (↓) the rows (vertical, column-wise)

> axis=1  →  Go ACROSS (→) the columns (horizontal, row-wise)

```text
     axis=1 →
   ┌─────────────┐
a  │ 1   2   3   │
x  │ 4   5   6   │
i  │ 7   8   9   │
s  └─────────────┘
=
0
↓
```
---


> **Example:**

```python

matrix = np.array([[10, 20, 30],
                   [40, 50, 60]])
#         Shape: (2, 3)  → 2 rows, 3 columns

# axis=0: Collapse ROWS (eliminate row dimension)
result = np.sum(matrix, axis=0)
print(result)  # [50, 70, 90]
print(result.shape)  # (3,)  ← Only columns left!

#     10  20  30
#   + 40  50  60
#   -----------
#     50  70  90


# axis=1: Collapse COLUMNS (eliminate column dimension)
result = np.sum(matrix, axis=1)
print(result)  # [60, 150]
print(result.shape)  # (2,)  ← Only rows left!

#   10 + 20 + 30 = 60
#   40 + 50 + 60 = 150
```

---
>axis=0 → Compare going DOWN ↓
         Result has COLUMN count (width stays)
         Tells you which ROW wins

>axis=1 → Compare going ACROSS →
         Result has ROW count (height stays)
         Tells you which COLUMN wins


````python
# returns the INDICES (positions) that would sort the array, NOT the sorted values themselves!
# Think of it as: "Tell me the ORDER of positions to pick elements to make it sorted"

np.argsort(random_array)
````

````python
# returns the INDEX (position) of the SMALLEST value in the array.
# covert it all in 1-d array

np.argmin(random_array)
````

````python
# returns the INDEX (position) of the LARGEST value in the array.
np.argmax(random_array)
````

````python
random_array
````

````python
# Finding MINIMUM in Each Row (argmin)

np.argmin(random_array, axis=1)
````

**How it works:**
```python

Row 0: [5, 0, 3, 3, 7]  →  Smallest is 0 at index 1  → Result: 1
        ↓  ↓  ↓  ↓  ↓
       [0][1][2][3][4]
           ↑ MIN!

Row 1: [9, 3, 5, 2, 4]  →  Smallest is 2 at index 3  → Result: 3
        ↓  ↓  ↓  ↓  ↓
       [0][1][2][3][4]
                 ↑ MIN!

Row 2: [7, 6, 8, 8, 1]  →  Smallest is 1 at index 4  → Result: 4
        ↓  ↓  ↓  ↓  ↓
       [0][1][2][3][4]
                    ↑ MIN!

RESULT: [1, 3, 4]
```

**How it works:**

```python

        Col 0   Col 1   Col 2   Col 3   Col 4
        ↓       ↓       ↓       ↓       ↓
Row 0:  5       0       3       3       7
Row 1:  9       3       5       2       4
Row 2:  7       6       8       8       1
        ↓       ↓       ↓       ↓       ↓
MIN:    5       0       3       2       1
        @row0   @row0   @row0   @row1   @row2

RESULT: [0, 0, 0, 1, 2]
```
