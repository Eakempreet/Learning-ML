````python
# Importing the modules
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt  # noqa: F401, F811
import numpy as np  # noqa: F401, F811
import sklearn
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score 
import pickle

# 0. Loading the data
heart_disease = pd.read_csv(r'..\Datasets\heart-disease.csv')

# 1. Getting the data ready
X = heart_disease.drop('target', axis=1)
y = heart_disease['target']

# 2. Choose the model and hyperparameters
# from sklearn.ensemble import RandomTreesEmbedding
clf = RandomForestClassifier()
clf.get_params()
# 3. Splitting the data into training data and testing data
# from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# 4. Fit the model with training data
clf.fit(X_train, y_train)

# 5. Evaluate the model on the training data and test data
## Testing on training data
y_preds = clf.predict(X_test)
clf.score(X_train, y_train)

## Testing on testing data
clf.score(X_test, y_test)

## Classification report 
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()

## Confusion matrix
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()

## Accuracy score
print("Accuracy score: ")
print(accuracy_score(y_true=y_test, y_pred=y_preds), "\n")

# 6. Improve a model
print("Improving model: ")
clf = RandomForestClassifier(n_estimators=90)  # trying out different params values
clf.fit(X_train, y_train)
y_preds = clf.predict(X_test)

clf.score(X_test, y_test)
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()
print("Accuracy Score \n")
print(accuracy_score(y_true=y_test, y_pred=y_preds))
print()

# 6. Save the model and reload it for future use
print("Saving the model...")
pickle.dump(clf, open("random_forest_model.pkl", "wb"))

print("Loading the model...")
pickle.load(open("random_forest_model.pkl", "rb"))
print("Voila!!! Done...")
````

### Understanding This Categorical Encoding Step

This is a **data preprocessing** step called **categorical encoding**.
The exact method being used here is **one-hot encoding**.

This cell is doing **two connected jobs**:

1. Build a **plan** for which columns should be converted.
2. Apply that plan to turn the data into **model-ready numbers**.

---

### The Big Idea

Your model can work with numbers, but not with text labels like:

- `Honda`
- `White`
- `Toyota`

So this cell acts like a **translator**.
It takes category labels and turns them into numeric columns the model can understand.

---

### Step-by-Step View

**Step 1: Pick the categorical columns**

These are the columns that contain labels rather than plain numeric measurements:

- `Make`
- `Colour`
- `Doors`

**Step 2: Create the converter**

`OneHotEncoder()` is the tool that turns each category into `0` and `1` columns.

**Step 3: Create the plan**

`ColumnTransformer(...)` is like a manager. It decides:

- which tool to use
- which columns should get that tool
- which columns should be left alone

**Step 4: Apply the plan**

`fit_transform(X)` means:

- `fit` = learn the possible categories from `X`
- `transform` = convert `X` using what it learned

---

### Visual Flow

```text
Original X
│
├─ Make         -> Honda, BMW, Toyota, Nissan
├─ Colour       -> White, Blue, Red, Green, Black
├─ Doors        -> 3, 4, 5
└─ Odometer     -> already numeric

ColumnTransformer Plan
│
├─ OneHotEncoder -> Make, Colour, Doors
└─ Passthrough   -> Odometer (KM)

Result: transformed_X
│
└─ all columns are now numeric
```

---

### Read This Part Like English

```python
("one_hot", one_hot, categorical_features)
```

| Piece | Meaning |
|---|---|
| `"one_hot"` | Just the name of this step |
| `one_hot` | The actual `OneHotEncoder()` tool |
| `categorical_features` | The columns to encode |

So this means:

> Create a step called `one_hot`, use the encoder tool, and apply it to `Make`, `Colour`, and `Doors`.

And this part:

```python
remainder="passthrough"
```

means:

> Keep any columns not listed above exactly as they are.


---

### Tiny Before and After Example

**Before**

| Make | Colour | Doors | Odometer (KM) |
|---|---|---:|---:|
| Honda | White | 4 | 35431 |

**After encoding**

| Make_BMW | Make_Honda | Make_Nissan | Make_Toyota | Colour_White | Doors_4 | Odometer (KM) |
|---:|---:|---:|---:|---:|---:|---:|
| 0 | 1 | 0 | 0 | 1 | 1 | 35431 |

The exact real output has more columns, but the idea is always the same:

- categories become `0` and `1` columns
- numeric columns can stay as they are

---

### Why Does It Become 13 Columns?

Because **one-hot encoding creates one new column for each category**.

From this dataset:

- `Make` has `4` categories: BMW, Honda, Nissan, Toyota
- `Colour` has `5` categories: Black, Blue, Green, Red, White
- `Doors` has `3` categories: 3, 4, 5
- `Odometer (KM)` is passed through as `1` numeric column

So the total is:

$$4 + 5 + 3 + 1 = 13$$

That is why `transformed_X` ends up with **13 columns**.

### One Real 13-Column Example

If one original row is:

| Make | Colour | Doors | Odometer (KM) |
|---|---|---:|---:|
| Honda | White | 4 | 35431 |

then the expanded 13-column version can look like this:

| Make_BMW | Make_Honda | Make_Nissan | Make_Toyota | Colour_Black | Colour_Blue | Colour_Green | Colour_Red | Colour_White | Doors_3 | Doors_4 | Doors_5 | Odometer (KM) |
|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|
| 0 | 1 | 0 | 0 | 0 | 0 | 0 | 0 | 1 | 0 | 1 | 0 | 35431 |

That table is the clearest way to see where the **13 columns** come from:

- `4` columns for `Make`
- `5` columns for `Colour`
- `3` columns for `Doors`
- `1` column for `Odometer (KM)`

---

### Most Important Takeaway

`transformer = ColumnTransformer(...)` means: **Here is the plan.**

`transformed_X = transformer.fit_transform(X)` means: **Now apply the plan to the data.**

- `X` stays as the original table
- `transformed_X` becomes the new numeric version the model can use

### Feature Scaling, for Now

**What is feature scaling?**

Feature scaling means resizing numeric features so they live on a similar scale.
Here, `Odometer (KM)` has very large values, so we scale it with `StandardScaler()`.

**Why use it here?**

- to preview how scaling is done in a real preprocessing workflow
- to stop large numeric values from dominating in models that care about scale
- to practice applying different preprocessing steps to different columns

**Why only `Odometer (KM)`?**

- `Make`, `Colour`, and `Doors` are handled with one-hot encoding
- `Odometer (KM)` is the main numeric feature, so it is the one being scaled

> In this section, scaling is mainly an **experiment / preview**. We will learn when it is truly needed in upcoming sections.

**Quick note:**
Tree-based models like `RandomForestRegressor` usually do **not** need feature scaling much, but trying it here helps you understand the workflow.

````python
# Fill missing values with Scikit-Learn
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

# Fill categorical values with "missing" and numerical values with mean
cat_imputer = SimpleImputer(strategy="constant",
                            fill_value="missing")
door_imputer = SimpleImputer(strategy="most_frequent")
num_imputer = SimpleImputer(strategy="mean")

# Define Columns
cat_features = ["Make", "Colour"]   # Catagorical features
door_feature = ["Doors"]
num_feature = ["Odometer (KM)"]

# Create an imputer (Something  that fills missing data)
imputer = ColumnTransformer([
    ("cat_imputer", cat_imputer, cat_features),
    ("door_imputer", door_imputer, door_feature),
    ("num_imputer", num_imputer, num_feature)
])

# Transform the data
filled_X_train = imputer.fit_transform(X_train)
filled_X_test = imputer.fit_transform(X_test)

# Check Filled X_train
filled_X_train


# Why we first split then inpute why not opposite it would have same time and effort?
# We split first to stop data leakage from the test set into training.
# In a proper workflow, the imputer learns the mean/mode from X_train only.
# Then the test set should use those same learned values, so evaluation stays honest.
````

````python
# Get our transformed data array's back into DataFrame's
car_sales_filled_train = pd.DataFrame(filled_X_train,
                                columns=["Make", "Colour", "Doors", "Odometer (KM)"])

car_sales_filled_test = pd.DataFrame(filled_X_test,
                                columns=["Make", "Colour", "Doors", "Odometer (KM)"])


# Check missing data in training set
car_sales_filled_train.isna().sum()


# Why creating two datasets why not just a single one and then do train test split?
````

##### **🌳 Decision Tree**

---

## What is it?

A decision tree makes predictions by asking a **series of yes/no questions**,
one after another, until it reaches a final answer.

---

## Real Life First

```
Should I carry an umbrella?

Is it cloudy?
├── NO  → Don't carry umbrella 🌤️
└── YES → Is rain forecasted?
               ├── NO  → Don't carry umbrella 🌥️
               └── YES → Carry umbrella ☂️
```

---

## In ML (Heart Disease Example)

```
Does this patient have heart disease?

Is age > 55?
├── NO  → Mostly healthy 🟢
└── YES → Is chest pain type = 3?
               ├── NO  → Mostly healthy 🟢
               └── YES → Likely sick 🔴
```

The tree finds the **best question** at each step by trying every
possible split and measuring which one separates the data most cleanly.

---

## The Vocabulary

```
             [Is age > 55?]          ← ROOT NODE   (first question)
             /             \
          [NO]             [YES]
           |                 |
       [Healthy]   [Is chest pain = 3?]  ← INTERNAL NODE (middle question)
                    /              \
                 [NO]             [YES]
                   |                |
              [Healthy] 🟢      [Sick] 🔴   ← LEAF NODES (final answer)
```

| Term          | Meaning                                      |
|---------------|----------------------------------------------|
| Root Node     | The very first question (best split overall) |
| Internal Node | Any question in the middle                   |
| Leaf Node     | Final answer — no more splitting             |
| Depth         | How many levels of questions the tree has    |

---

## How Does It Pick Questions?

### Purity — the core idea

Think of bags of red and blue balls:

```
Bag A:  50 red, 50 blue  →  Maximum confusion   (IMPURE)
Bag B:  90 red, 10 blue  →  Mostly guessable    (LESS PURE)
Bag C: 100 red,  0 blue  →  Perfect, always know (PURE)
```

**The tree wants to create bags that look like Bag C.**

At every step it asks: *"Which question splits my data into the purest bags?"*

### The measurement — Gini Impurity

| Gini = 0   | Perfectly pure  (all one class)     |
|------------|-------------------------------------|
| Gini = 0.5 | Maximally impure (50/50 split)      |

The tree tries **every possible question on every feature**,
calculates Gini for each split, and picks the one that
**lowers Gini the most**.


---

## For Regression (Predicting Numbers)

Same structure. Same splitting logic.
But the leaf doesn't say "sick/healthy" —
it says the **average value** of all training samples that ended up there.

---

## The Big Problem — Overfitting

A single tree, if grown too deep, **memorizes** the training data.

```python
clf.score(X_train, y_train)  # → 1.0   (100% — memorized)
clf.score(X_test, y_test)    # → 0.85  (drops on new data)
```

The gap between training and test score = **overfitting**.

Like a student who memorizes past exam papers word for word —
then fails when the actual exam has slightly different questions.

> ✅ This is exactly the problem **Random Forest** was invented to solve.

---

## One Line Summary

> A decision tree asks the most useful yes/no question at each step,
> measured by how purely it separates the data,
> until the answer is clear enough to stop.

## 🌲 Random Forest — Intuition

---

### 🌳 Why One Tree Is Not Enough

A single decision tree, grown deep enough, **memorizes** the training data.
It fits perfectly on what it has seen — but struggles on new data.
This is called **overfitting**.

```python
clf.score(X_train, y_train)  # → 1.0   (memorized)
clf.score(X_test, y_test)    # → 0.85  (drops on new data)
```

> **One tree is too opinionated.**
> It saw one version of the data and built one rigid structure.

---

### 🗳️ The Solution — Ask Many Trees

> Don't rely on one tree. Build hundreds of trees and let them **vote**.

This is the same logic as asking many friends for advice.
No single friend is always right — but the **majority opinion** tends to be far more accurate.

---

### 🎲 Where the "Random" Comes From

If all trees trained on the same data → they'd all build the same tree.
Voting 100 identical trees would be pointless.

So Random Forest introduces **two deliberate sources of randomness.**

---

**Randomness 1 — Different Data for Each Tree (Bootstrapping)**

Each tree gets a **random sample** of the training data, with replacement.
The same row can appear multiple times. Some rows won't appear at all.

```
Full data:    [P1, P2, P3, P4, P5, P6, P7, P8, P9, P10]

Tree 1 sees:  [P1, P1, P3, P5, P5, P7, P8, P10, P10, P2]
Tree 2 sees:  [P2, P4, P4, P6, P7, P7, P9,  P1,  P3,  P8]
Tree 3 sees:  [P3, P5, P6, P6, P8, P9, P10, P2,  P4,  P7]
```

Every tree trains on a slightly different version of reality.
So every tree builds a slightly different structure.

#### **Bootstarpping**
- Bootstrapping means: 
  - Make a new dataset by randomly picking rows from your original dataset with replacement.

- “With replacement” means:
  - After you pick a row, you put it back before the next pick.
    - the same row can appear multiple times,
    - some rows might not appear at all.
---

**Randomness 2 — Different Features at Each Split**

At every split, the algorithm doesn't consider **all** features.
It picks a **random subset** — controlled by `max_features='sqrt'`.

```
All features available → [age, cp, chol, thalach, slope, ca, thal ...]
Random subset chosen   → [age, chol, thal]
Best question picked from ONLY those 3
```

No single feature can dominate every tree.

---

### 🗳️ How the Voting Works

**Classification** → majority vote

```
Tree 1  →  Heart Disease ✓
Tree 2  →  Heart Disease ✓
Tree 3  →  No Disease    ✗
Tree 4  →  Heart Disease ✓
Tree 5  →  Heart Disease ✓

Final   →  Heart Disease  (4 out of 5)
```

**Regression** → average of all predictions

```
Tree 1  →  $320,000
Tree 2  →  $305,000
Tree 3  →  $340,000

Final   →  $321,667
```

---

### ✅ Why Overfitting Gets Fixed

Each tree still overfits — but to a **different slice of the data**.

When they vote:
- The **noise cancels out** — each tree's noise is different
- The **signal survives** — real patterns appear in most trees

---

### 🗺️ The Full Picture

```
            Original Training Data
                     ↓
       ┌─────────────┼─────────────┐
       ↓             ↓             ↓
 Random Sample  Random Sample  Random Sample  ...
       ↓             ↓             ↓
     Tree 1        Tree 2        Tree 3        ...
 (rand features)(rand features)(rand features)
       ↓             ↓             ↓
 Prediction 1  Prediction 2  Prediction 3
          ↘         ↓         ↙
             Vote / Average
                  ↓
           Final Prediction
```

---

### ⚙️ Hyperparameters Decoded

| Hyperparameter     | What it controls                                 |
|--------------------|--------------------------------------------------|
| `n_estimators=100` | How many trees to build                          |
| `max_depth`        | How deep each individual tree can grow           |
| `max_features`     | How many features considered at each split       |
| `bootstrap=True`   | Whether to use random sampling w/ replacement    |
| `random_state`     | Fixes randomness so results are reproducible     |

---

### 💡 One Line Summary

> Random Forest builds many decision trees, each trained on slightly different data
> and features, then combines their answers — so individual errors cancel out
> and the final prediction is far more reliable than any single tree.

- **Classifier** → predicts categories → `RandomForestClassifier`
- **Regressor**  → predicts numbers   → `RandomForestRegressor`

##### What is R² (Coefficient of Determination)?
> Returned by `model.score()` for regression models

---

### 1. The Main Idea

R² answers one question:

> How much better is my model than a very dumb guess?

That dumb guess is:

> Always predict the average of the target values.

---

### 2. Start With the Baseline

Imagine we are predicting house prices.

```text
Actual prices:  $300k   $350k   $400k   $450k
```

First, find the average:

```text
Average = (300 + 350 + 400 + 450) / 4
        = $375k
```

So the baseline prediction is:

```text
For every house, guess $375k
```

This is the reference point R² compares against.

---

### 3. Baseline Errors

Now check how wrong that baseline is.

```text
House 1: actual $300k, guessed $375k  -> error = 75k
House 2: actual $350k, guessed $375k  -> error = 25k
House 3: actual $400k, guessed $375k  -> error = 25k
House 4: actual $450k, guessed $375k  -> error = 75k
```

In R², we square the errors:

```text
(75)^2 + (25)^2 + (25)^2 + (75)^2
= 5625 + 625 + 625 + 5625
= 12500
```

This is the baseline total error.

---

### 4. Model Guesses

Now suppose your model predicts:

```text
House 1 -> $310k
House 2 -> $360k
House 3 -> $395k
House 4 -> $445k
```

Now compute the model's errors:

```text
House 1: actual $300k, predicted $310k  -> error = 10k
House 2: actual $350k, predicted $360k  -> error = 10k
House 3: actual $400k, predicted $395k  -> error = 5k
House 4: actual $450k, predicted $445k  -> error = 5k
```

Square them:

```text
(10)^2 + (10)^2 + (5)^2 + (5)^2
= 100 + 100 + 25 + 25
= 250
```

This is the model's total error.

---

### 5. The Formula

```text
R² = 1 - (model error / baseline error)
```

Plug in the numbers:

```text
R² = 1 - (250 / 12500)
   = 1 - 0.02
   = 0.98
```

Meaning:

> The model explains 98% of the variation in house prices.

---

### 6. How to Read R²

| R² value | Meaning |
|---|---|
| `1.0` | Perfect predictions |
| `0.8` to `0.99` | Very strong model |
| `0.5` to `0.79` | Useful, but not great |
| `0.0` | No better than guessing the average |
| Negative | Worse than guessing the average |

---

### 7. Your Ridge Model Score

Your notebook showed:

```text
R² = 0.5758
```

That means:

1. Your model is better than the baseline.
2. It explains about 57.58% of the variation in the target.
3. About 42.42% is still unexplained.

That unexplained part could come from:

- noise
- missing features
- imperfect model choice
- limited data

---

### 8. R² vs Accuracy

- **For regression:**

```python
model.score(X_test, y_test)
```

returns:

```text

```

- **For classification:**

```python

```

returns:

```text
Accuracy = correct predictions / total predictions
```

So both use `.score()`, but they measure different things.

Make predictions with `predicit_proba()`

#### **Predict vs Predict_Proba: Final Decision vs Model Confidence**

When working with a **classification model**, these two methods answer **different questions**:

| Method | Main question it answers | What it returns | Example output | Best use |
|---|---|---|---|---|
| `predict()` | Which class should the model choose? | Final class label | `[1, 0, 1]` | When you want the direct prediction |
| `predict_proba()` | How confident is the model about each class? | Probability for every class | `[[0.20, 0.80], [0.91, 0.09], [0.35, 0.65]]` | When you want confidence and uncertainty |

---

#### 1. `predict()` -> the model's final answer

This method gives the **chosen class label** directly.

For a heart disease example:

- `0` = No heart disease
- `1` = Heart disease

```python
clf.predict(X_test[:3])
# array([1, 0, 1])
```

That means the model's final decisions are:

- first patient -> class `1`
- second patient -> class `0`
- third patient -> class `1`

Think of `predict()` as the model saying:

> "This is my final choice."

---

#### 2. `predict_proba()` -> the model's confidence for each class

This method gives the **probability of every possible class**.

```python
clf.predict_proba(X_test[:3])
# array([
#   [0.20, 0.80],
#   [0.91, 0.09],
#   [0.35, 0.65]
# ])
```

Read each row carefully:

- `[0.20, 0.80]` means 20% chance of class `0`, 80% chance of class `1`
- `[0.91, 0.09]` means 91% chance of class `0`, 9% chance of class `1`
- `[0.35, 0.65]` means 35% chance of class `0`, 65% chance of class `1`

Think of `predict_proba()` as the model saying:

> "Here is how confident I am about each possible answer."

---

#### 3. How they are connected

In many classification models, `predict()` chooses the class with the **highest probability** from `predict_proba()`.

Example:

- `[0.20, 0.80]` -> choose class `1`
- `[0.91, 0.09]` -> choose class `0`
- `[0.35, 0.65]` -> choose class `1`

So a simple memory trick is:

- `predict_proba()` = full confidence picture
- `predict()` = final selected answer

---

#### 4. Real-world analogy

Imagine a doctor reviewing a patient's test results:

- `predict()` = "The patient has heart disease."
- `predict_proba()` = "There is an 80% chance of heart disease and a 20% chance of no heart disease."

One gives the **decision**.
The other gives the **confidence behind the decision**.

---

#### 5. Most important takeaway

Use `predict()` when you only need the final class.
Use `predict_proba()` when you also want to understand **how certain** the model is.


> Short version: `predict()` tells you **what** the model chose. `predict_proba()` tells you **how sure** it was.

*How the cross-validation working*

#### **Cross-Validation (CV)**

> **Goal:** get a *more trustworthy* estimate of how your model will do on **new/unseen data**.

> **Important:** CV does **not** magically make the model better — it makes the **evaluation** less dependent on luck.

---
### 1) Why do we need it? ("lucky" vs "unlucky")
If you do **one** train/test split, your score can change a lot depending on *which* rows ended up in the test set:
- **Lucky split**: test set happens to be easier → score looks **too high**
- **Unlucky split**: test set happens to be harder → score looks **too low**

> Analogy: tasting soup with one spoon. One scoop might be salty (unlucky) or bland (lucky). Multiple spoonfuls is fairer.

---
### 2) What is cross-validation?
**Cross-validation** = run several train/test experiments (in a structured way) and collect the scores.
Then you usually look at the **average** score (and how much the scores vary).

> Think: **multiple mini-tests** instead of one final test.

---
### 3) How does 5-fold CV work? (`cv=5`)
With `cv=5`, your dataset is split into **5 parts** (called *folds*):

> `[Fold 1] [Fold 2] [Fold 3] [Fold 4] [Fold 5]`

> You do 5 rounds: each fold becomes the test set once.

- Round 1: train on folds 2–5, test on fold 1
- Round 2: train on folds 1,3–5, test on fold 2
- Round 3: train on folds 1–2,4–5, test on fold 3
- Round 4: train on folds 1–3,5, test on fold 4
- Round 5: train on folds 1–4, test on fold 5

> Result: you get **5 scores** (one score per round).

> For classification, scikit-learn usually tries to keep class balance similar in each fold.

---
### 4) What does `cross_val_score` return?
```python
cross_val_score(clf, X, y)
```
This returns an array of scores, like:
`array([0.82, 0.79, 0.85, 0.80, 0.83])`

> The model was tested 5 times, so you got 5 scores.

Common summary:
```python
scores = cross_val_score(clf, X, y, cv=5)
scores.mean()  # average performance
scores.std()   # stability (lower = more consistent)
```

> If the scores vary a lot, it means performance depends heavily on which data slice you tested on.

---
### 5) Why these two lines look the same
```python
cross_val_score(clf, X, y)
```
vs
```python
cross_val_score(clf, X, y, cv=5)
```
In scikit-learn, the default is usually **5 folds**, so these are often effectively the same.

> Writing `cv=5` is just **more explicit** and easier to remember later.

---
### 6) CV vs a single train/test split (quick comparison)
- **One split**: faster, but can be noisy (lucky/unlucky)
- **Cross-validation**: slower, but gives a more reliable estimate

> Memory sentence: **CV is multiple mini-tests, so your score is less about luck.**

#### **Default Scoring in `cross_val_score`**

When you run cross-validation **without** setting `scoring`, sklearn uses the model's **own default `.score()`** method.

So this:

```python
cross_val_score(model, X, y, cv=5)
```

means:
- "Use 5-fold CV"
- "For each fold, score using the model's default scoring rule"

**Common defaults:**
- **Classifier** (e.g., `RandomForestClassifier`) -> `.score()` = **Accuracy**
- **Regressor** (e.g., `RandomForestRegressor`) -> `.score()` = **R²**

So people say: **default CV scoring = estimator default scoring**.

If you want a different metric, set it explicitly, for example:

```python
cross_val_score(clf, X, y, cv=5, scoring="f1")
```
