````python
# Importing the modules
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt  # noqa: F401, F811
import numpy as np  # noqa: F401, F811
import sklearn
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score 
import pickle

# 0. Loading the data
heart_disease = pd.read_csv(r'..\Datasets\heart-disease.csv')

# 1. Getting the data ready
X = heart_disease.drop('target', axis=1)
y = heart_disease['target']

# 2. Choose the model and hyperparameters
# from sklearn.ensemble import RandomTreesEmbedding
clf = RandomForestClassifier()
clf.get_params()
# 3. Splitting the data into training data and testing data
# from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# 4. Fit the model with training data
clf.fit(X_train, y_train)

# 5. Evaluate the model on the training data and test data
## Testing on training data
y_preds = clf.predict(X_test)
clf.score(X_train, y_train)

## Testing on testing data
clf.score(X_test, y_test)

## Classification report 
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()

## Confusion matrix
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()

## Accuracy score
print("Accuracy score: ")
print(accuracy_score(y_true=y_test, y_pred=y_preds), "\n")

# 6. Improve a model
print("Improving model: ")
clf = RandomForestClassifier(n_estimators=90)  # trying out different params values
clf.fit(X_train, y_train)
y_preds = clf.predict(X_test)

clf.score(X_test, y_test)
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()
print("Accuracy Score \n")
print(accuracy_score(y_true=y_test, y_pred=y_preds))
print()

# 6. Save the model and reload it for future use
print("Saving the model...")
pickle.dump(clf, open("random_forest_model.pkl", "wb"))

print("Loading the model...")
pickle.load(open("random_forest_model.pkl", "rb"))
print("Voila!!! Done...")
````

### Understanding This Categorical Encoding Step

This is a **data preprocessing** step called **categorical encoding**.
The exact method being used here is **one-hot encoding**.

This cell is doing **two connected jobs**:

1. Build a **plan** for which columns should be converted.
2. Apply that plan to turn the data into **model-ready numbers**.

---

### The Big Idea

Your model can work with numbers, but not with text labels like:

- `Honda`
- `White`
- `Toyota`

So this cell acts like a **translator**.
It takes category labels and turns them into numeric columns the model can understand.

---

### Step-by-Step View

**Step 1: Pick the categorical columns**

These are the columns that contain labels rather than plain numeric measurements:

- `Make`
- `Colour`
- `Doors`

**Step 2: Create the converter**

`OneHotEncoder()` is the tool that turns each category into `0` and `1` columns.

**Step 3: Create the plan**

`ColumnTransformer(...)` is like a manager. It decides:

- which tool to use
- which columns should get that tool
- which columns should be left alone

**Step 4: Apply the plan**

`fit_transform(X)` means:

- `fit` = learn the possible categories from `X`
- `transform` = convert `X` using what it learned

---

### Visual Flow

```text
Original X
│
├─ Make         -> Honda, BMW, Toyota, Nissan
├─ Colour       -> White, Blue, Red, Green, Black
├─ Doors        -> 3, 4, 5
└─ Odometer     -> already numeric

ColumnTransformer Plan
│
├─ OneHotEncoder -> Make, Colour, Doors
└─ Passthrough   -> Odometer (KM)

Result: transformed_X
│
└─ all columns are now numeric
```

---

### Read This Part Like English

```python
("one_hot", one_hot, categorical_features)
```

| Piece | Meaning |
|---|---|
| `"one_hot"` | Just the name of this step |
| `one_hot` | The actual `OneHotEncoder()` tool |
| `categorical_features` | The columns to encode |

So this means:

> Create a step called `one_hot`, use the encoder tool, and apply it to `Make`, `Colour`, and `Doors`.

And this part:

```python
remainder="passthrough"
```

means:

> Keep any columns not listed above exactly as they are.


---

### Tiny Before and After Example

**Before**

| Make | Colour | Doors | Odometer (KM) |
|---|---|---:|---:|
| Honda | White | 4 | 35431 |

**After encoding**

| Make_BMW | Make_Honda | Make_Nissan | Make_Toyota | Colour_White | Doors_4 | Odometer (KM) |
|---:|---:|---:|---:|---:|---:|---:|
| 0 | 1 | 0 | 0 | 1 | 1 | 35431 |

The exact real output has more columns, but the idea is always the same:

- categories become `0` and `1` columns
- numeric columns can stay as they are

---

### Why Does It Become 13 Columns?

Because **one-hot encoding creates one new column for each category**.

From this dataset:

- `Make` has `4` categories: BMW, Honda, Nissan, Toyota
- `Colour` has `5` categories: Black, Blue, Green, Red, White
- `Doors` has `3` categories: 3, 4, 5
- `Odometer (KM)` is passed through as `1` numeric column

So the total is:

$$4 + 5 + 3 + 1 = 13$$

That is why `transformed_X` ends up with **13 columns**.

### One Real 13-Column Example

If one original row is:

| Make | Colour | Doors | Odometer (KM) |
|---|---|---:|---:|
| Honda | White | 4 | 35431 |

then the expanded 13-column version can look like this:

| Make_BMW | Make_Honda | Make_Nissan | Make_Toyota | Colour_Black | Colour_Blue | Colour_Green | Colour_Red | Colour_White | Doors_3 | Doors_4 | Doors_5 | Odometer (KM) |
|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|
| 0 | 1 | 0 | 0 | 0 | 0 | 0 | 0 | 1 | 0 | 1 | 0 | 35431 |

That table is the clearest way to see where the **13 columns** come from:

- `4` columns for `Make`
- `5` columns for `Colour`
- `3` columns for `Doors`
- `1` column for `Odometer (KM)`

---

### Most Important Takeaway

`transformer = ColumnTransformer(...)` means: **Here is the plan.**

`transformed_X = transformer.fit_transform(X)` means: **Now apply the plan to the data.**

- `X` stays as the original table
- `transformed_X` becomes the new numeric version the model can use

### Feature Scaling, for Now

**What is feature scaling?**

Feature scaling means resizing numeric features so they live on a similar scale.
Here, `Odometer (KM)` has very large values, so we scale it with `StandardScaler()`.

**Why use it here?**

- to preview how scaling is done in a real preprocessing workflow
- to stop large numeric values from dominating in models that care about scale
- to practice applying different preprocessing steps to different columns

**Why only `Odometer (KM)`?**

- `Make`, `Colour`, and `Doors` are handled with one-hot encoding
- `Odometer (KM)` is the main numeric feature, so it is the one being scaled

> In this section, scaling is mainly an **experiment / preview**. We will learn when it is truly needed in upcoming sections.

**Quick note:**
Tree-based models like `RandomForestRegressor` usually do **not** need feature scaling much, but trying it here helps you understand the workflow.

````python
# Drop the rows with no labels
car_sales_missing.dropna(subset=["Price"], inplace=True)
car_sales_missing.isna().sum()
````

````python
# Split into X & y
X = car_sales_missing.drop("Price", axis=1)
y = car_sales_missing["Price"]

# Split data into train and test
np.random.seed(42)
X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y,
                                                    test_size=0.2)
````

````python
# Fill missing values with Scikit-Learn
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

# Fill categorical values with "missing" and numerical values with mean
cat_imputer = SimpleImputer(strategy="constant",
                            fill_value="missing")
door_imputer = SimpleImputer(strategy="most_frequent")
num_imputer = SimpleImputer(strategy="mean")

# Define Columns
cat_features = ["Make", "Colour"]   # Catagorical features
door_feature = ["Doors"]
num_feature = ["Odometer (KM)"]

# Create an imputer (Something  that fills missing data)
imputer = ColumnTransformer([
    ("cat_imputer", cat_imputer, cat_features),
    ("door_imputer", door_imputer, door_feature),
    ("num_imputer", num_imputer, num_feature)
])

# Transform the data
filled_X_train = imputer.fit_transform(X_train)
filled_X_test = imputer.fit_transform(X_test)

# Check Filled X_train
filled_X_train


# Why we first split then inpute why not opposite it would have same time and effort?
# We split first to stop data leakage from the test set into training.
# In a proper workflow, the imputer learns the mean/mode from X_train only.
# Then the test set should use those same learned values, so evaluation stays honest.
````

````python
# Get our transformed data array's back into DataFrame's
car_sales_filled_train = pd.DataFrame(filled_X_train,
                                columns=["Make", "Colour", "Doors", "Odometer (KM)"])

car_sales_filled_test = pd.DataFrame(filled_X_test,
                                columns=["Make", "Colour", "Doors", "Odometer (KM)"])


# Check missing data in training set
car_sales_filled_train.isna().sum()


# Why creating two datasets why not just a single one and then do train test split?
````

````python
# Check to see the original... still missing values
car_sales_missing.isna().sum()
````

````python
# Now let's one hot encode the features with the same code as before 
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = ["Make", "Colour", "Doors"]
one_hot = OneHotEncoder()

transformer = ColumnTransformer([("one_hot",
                                  one_hot,
                                  categorical_features)],
                                  remainder="passthrough")

# Fill train and test values separately
transformed_X_train = transformer.fit_transform(car_sales_filled_train)
transformed_X_test = transformer.fit_transform(car_sales_filled_test)

transformed_X_train.toarray()

# Why again transform the traning sets differently why just normally transform the X?
# In a proper workflow, the encoder learns categories from the training set only.
# Then the test set should be transformed with that same learned mapping.
# This keeps train and test columns aligned and avoids leakage.
````

````python
# Now we've transformed X, let's see if we can fit a model
np.random.seed(42)
from sklearn.ensemble import RandomForestRegressor  # noqa: E402

model = RandomForestRegressor()

# Make sure to use transformed (filled and one-hot encoded X data)
model.fit(transformed_X_train, y_train)
model.score(transformed_X_test, y_test)
````

### Option 1 vs Option 2 (Quick Guide)

| Part | Option 1: pandas fill | Option 2: scikit-learn transformers |
|---|---|---|
| Main idea | Fill missing values manually with pandas (`fillna`, `dropna`) | Use `SimpleImputer` + `OneHotEncoder` + `ColumnTransformer` |
| Best for | Learning basics, quick experiments, understanding each step | Real ML workflow, cleaner validation, reusable preprocessing |
| Pros | Very easy to read and debug | More robust and scalable for ML projects |
| Cons | Easy to make leakage mistakes if split order is wrong | Slightly more setup and easier to misuse at first |
| Reuse later | Manual repetition needed | Can be reused in pipelines for future data |

---

### Which one is recommended?

- Use **Option 1** when you are learning concepts and want to see each operation clearly.
- Use **Option 2** for proper model training and real projects.

In short: **Option 1 is beginner-friendly, Option 2 is production-friendly.**

---

### Important rule for Option 2

- `fit_transform(...)` on **training** data
- `transform(...)` on **test** data

This avoids data leakage and gives a more honest model score.

---

### Why scores can differ between Option 1 and Option 2

- Option 2 may train on fewer rows after dropping missing target (`Price`) rows.
- Imputed values are estimates, so they can add a little noise.
- Small workflow differences (especially leakage mistakes) can change scores noticeably.

  - Sometimes you'll see `clf` (short for classifier) used as a classification estimator

#### **Quick Note: Algorithm vs Estimator vs Model**

1. **Algorithm** = the learning idea (example: Random Forest).
2. **Estimator** = the sklearn object that implements that idea (example: `RandomForestClassifier()`).
3. **Model** = that estimator after training with `.fit(X, y)`.


> In casual conversation, people often use these words interchangeably.

---

**Example:**

```python
# You have an algorithm (Random Forest) in mind
# You create an estimator from it
clf = RandomForestClassifier(n_estimators=100)

# clf is an ESTIMATOR — knows the rules, seen no data
# You cannot call clf.predict() here — it will throw an error

# You train it
clf.fit(X_train, y_train)

# clf is now a MODEL — it has learned from data
# Now you can predict
clf.predict(X_test)
```

##### **What is Coefficient of Determination (R²)?**
> Given by model.score()
---

##### 🎯 **Simple Explanation**

Imagine you're predicting house prices.

**Without a model:** You just guess the average price for every house = **bad**  
**With a model:** You use features (size, location, age) to make smart guesses = **better**

**R² answers:** "How much better is my smart model compared to just guessing the average?"

---

##### 📊 **Visual Story**

```
Actual prices:     $300k,  $350k,  $400k,  $450k
Average price:     $375k   (if you always guess this)
Your model guesses: $310k,  $360k,  $395k,  $445k

R² = "How much of the variation did your model explain?"
```

**Without model:** Price varies $75k from average  
**Your model error:** Much smaller variations  
**R²:** Percentage of that variation your model captured

---

##### 🧮 **The Formula (Explained Simply)**

```python
R² = 1 - (Errors from your model) / (Errors from guessing average)

# If model is perfect:
R² = 1 - 0 = 1.0  ✓✓✓

# If model is as bad as guessing average:
R² = 1 - 1 = 0.0  ✗✗✗

# If model is halfway good:
R² = 1 - 0.5 = 0.5  ✓✗
```

---

##### 📈 **What Different R² Scores Mean**

| R² Score | Meaning | Performance |
|---|---|---|
| **1.0 (100%)** | Perfect predictions | Model explains ALL variation ⭐⭐⭐ |
| **0.9 (90%)** | Excellent | Model explains 90% of variation ⭐⭐ |
| **0.7 (70%)** | Good | Model explains 70% of variation ⭐ |
| **0.5 (50%)** | Okay | Model explains only half ⚠️ |
| **0.3 (30%)** | Poor | Model explains 30%, lots of randomness 😕 |
| **0.0 (0%)** | Useless | Model is as bad as guessing average ❌ |
| **Negative** | Terrible | Model is WORSE than guessing average 💥 |

---

##### 🔍 **Real Example: House Prices**

Your Ridge model scored: **0.5758** (57.58%)

**What this means:**
- ✅ Your model explains **57.58%** of house price changes
- ❌ **42.42%** of price changes are due to factors your model doesn't know
- 🤔 Your model is pretty good but not perfect

**In plain English:**  
"If house prices vary $1,000 from the average, your model explains about $576 of that variation. The other $424 is random or from unknown factors."

---

##### 🎬 **Why Not Just Use Accuracy?**

**Accuracy** (used for classification):
- "Out of 100 predictions, how many were correct?"
- Example: 87/100 correct = 87% accuracy

**R² (used for regression):**
- "What percentage of the variation in prices did I explain?"
- Example: Model explains 57.58% of price variations = R² = 0.5758

💡 **Key Difference:**  
- Accuracy = Count of right/wrong  
- R² = Quality of prediction relative to just guessing average

---

##### 🛠️ **How to Use R² in Your Code**

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split

model = Ridge()
model.fit(X_train, y_train)

# This line gives you R²
r_squared = model.score(X_test, y_test)
print(f"R² Score: {r_squared:.4f}")  # Output: 0.5758

# Interpret it
if r_squared > 0.8:
    print("Excellent model! ⭐⭐⭐")
elif r_squared > 0.6:
    print("Good model! ⭐⭐")
elif r_squared > 0.4:
    print("Okay model, could be better. ⭐")
else:
    print("Poor model, needs improvement. ❌")
```

---

##### 🎓 **Advanced Note**

**Formula breakdown:**

```python
SS_res = Σ(actual - predicted)²      # Sum of squared errors from model
SS_tot = Σ(actual - average)²        # Sum of squared errors from guessing average

R² = 1 - (SS_res / SS_tot)

# Example:
# SS_res = 100  (model's total error)
# SS_tot = 234  (average's total error)
# R² = 1 - (100/234) = 1 - 0.427 = 0.573 = 57.3%
```

---

##### ✅ **Remember This**

- **R² tells you what % of variation your model explains**
- **R² = 1.0** = Perfect (explains 100%)
- **R² = 0.0** = Useless (explains 0%, as bad as average)
- **R² can be negative** = Model is worse than just guessing average (scary! 😱)
- **For California Housing:** Your Ridge model is okay (57.58%), room for improvement
