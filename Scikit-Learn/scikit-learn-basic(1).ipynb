## **0. An end-to-end Scikit-Learn Workflow**

````python
# 1. Get the data ready

import pandas as pd  # noqa: F811
heart_disease = pd.read_csv(r'..\Datasets\heart-disease.csv')
heart_disease.head()
````

````python
# Create X (Feature matrix)
X = heart_disease.drop('target', axis=1)

# Create Y (Label matrix/ Labels)
y = heart_disease['target']

# Expected output (classification)
````

````python
## **2. Choose the right model and hyperparameters**

from sklearn.ensemble import RandomForestClassifier

# Creates a new ML model and naming it clf (classifier)
clf = RandomForestClassifier()

# We'll Keep the default hyperparameters
clf.get_params()

# clf.get_params()
# ↑ Shows ALL those default hyperparameters
````

HYPERPARAMETERS:
───────────────────────────────────────
- Settings you choose BEFORE training
- They control HOW the model learns
- You set them, the model uses them

---

`clf` is a MODEL ← An untrained RandomForest

- Has hyperparameters set to defaults
- Ready to be trained
- Waiting for data
- Can't predict yet (needs training first!)

````python
# 3. Fit the model to the training data

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Randomly splits your data into 2 groups:
# 80% to train the model,
# 20% to test if it actually learned (0.2)
````

```text

│        │        └─ Training targets (answers for learning)
│        └─ Testing features (unseen data to test on)

```

````python
clf.fit(X_train, y_train)

# Trains the model by showing it the features (X_train)
# and their correct answers (y_train),
# so it learns the patterns
````

````python
# Makes predictions on NEW, UNSEEN data

y_preds = clf.predict(X_test)
y_preds
````

````python
# 4. Evaluate the model on the training data and test data

# Training Accuracy
clf.score(X_train, y_train)
````

**Means:** Model got 100% correct on the data it learned from!

````python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print(classification_report(y_test, y_preds))
# What it does: 
# Gives you a FULL report with multiple metrics
````

### **Summary**

````python
# Importing the modules
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt  # noqa: F401, F811
import numpy as np  # noqa: F401, F811
import sklearn
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score 
import pickle

# 0. Loading the data
heart_disease = pd.read_csv(r'..\Datasets\heart-disease.csv')

# 1. Getting the data ready
X = heart_disease.drop('target', axis=1)
y = heart_disease['target']

# 2. Choose the model and hyperparameters
# from sklearn.ensemble import RandomTreesEmbedding
clf = RandomForestClassifier()
clf.get_params()
# 3. Splitting the data into training data and testing data
# from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# 4. Fit the model with training data
clf.fit(X_train, y_train)

# 5. Evaluate the model on the training data and test data
## Testing on training data
y_preds = clf.predict(X_test)
clf.score(X_train, y_train)

## Testing on testing data
clf.score(X_test, y_test)

## Classification report 
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()

## Confusion matrix
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()

## Accuracy score
print("Accuracy score: ")
print(accuracy_score(y_true=y_test, y_pred=y_preds), "\n")

# 6. Improve a model
print("Improving model: ")
clf = RandomForestClassifier(n_estimators=90)  # trying out different params values
clf.fit(X_train, y_train)
y_preds = clf.predict(X_test)

clf.score(X_test, y_test)
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()
print("Accuracy Score \n")
print(accuracy_score(y_true=y_test, y_pred=y_preds))
print()

# 6. Save the model and reload it for future use
print("Saving the model...")
pickle.dump(clf, open("random_forest_model.pkl", "wb"))

print("Loading the model...")
pickle.load(open("random_forest_model.pkl", "rb"))
print("Voila!!! Done...")
````

- `conda update --all` updates all packages in your active environment.

**3. Search and check a package**
