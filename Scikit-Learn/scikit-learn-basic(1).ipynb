````python
# Importing the modules
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt  # noqa: F401, F811
import numpy as np  # noqa: F401, F811
import sklearn
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score 
import pickle

# 0. Loading the data
heart_disease = pd.read_csv(r'..\Datasets\heart-disease.csv')

# 1. Getting the data ready
X = heart_disease.drop('target', axis=1)
y = heart_disease['target']

# 2. Choose the model and hyperparameters
# from sklearn.ensemble import RandomTreesEmbedding
clf = RandomForestClassifier()
clf.get_params()
# 3. Splitting the data into training data and testing data
# from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# 4. Fit the model with training data
clf.fit(X_train, y_train)

# 5. Evaluate the model on the training data and test data
## Testing on training data
y_preds = clf.predict(X_test)
clf.score(X_train, y_train)

## Testing on testing data
clf.score(X_test, y_test)

## Classification report 
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()

## Confusion matrix
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()

## Accuracy score
print("Accuracy score: ")
print(accuracy_score(y_true=y_test, y_pred=y_preds), "\n")

# 6. Improve a model
print("Improving model: ")
clf = RandomForestClassifier(n_estimators=90)  # trying out different params values
clf.fit(X_train, y_train)
y_preds = clf.predict(X_test)

clf.score(X_test, y_test)
print("Classification report: \n")
print(classification_report(y_test, y_preds))
print()
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_preds))
print()
print("Accuracy Score \n")
print(accuracy_score(y_true=y_test, y_pred=y_preds))
print()

# 6. Save the model and reload it for future use
print("Saving the model...")
pickle.dump(clf, open("random_forest_model.pkl", "wb"))

print("Loading the model...")
pickle.load(open("random_forest_model.pkl", "rb"))
print("Voila!!! Done...")
````

### Understanding This Categorical Encoding Step

This is a **data preprocessing** step called **categorical encoding**.
The exact method being used here is **one-hot encoding**.

This cell is doing **two connected jobs**:

1. Build a **plan** for which columns should be converted.
2. Apply that plan to turn the data into **model-ready numbers**.

---

### The Big Idea

Your model can work with numbers, but not with text labels like:

- `Honda`
- `White`
- `Toyota`

So this cell acts like a **translator**.
It takes category labels and turns them into numeric columns the model can understand.

---

### Step-by-Step View

**Step 1: Pick the categorical columns**

These are the columns that contain labels rather than plain numeric measurements:

- `Make`
- `Colour`
- `Doors`

**Step 2: Create the converter**

`OneHotEncoder()` is the tool that turns each category into `0` and `1` columns.

**Step 3: Create the plan**

`ColumnTransformer(...)` is like a manager. It decides:

- which tool to use
- which columns should get that tool
- which columns should be left alone

**Step 4: Apply the plan**

`fit_transform(X)` means:

- `fit` = learn the possible categories from `X`
- `transform` = convert `X` using what it learned

---

### Visual Flow

```text
Original X
│
├─ Make         -> Honda, BMW, Toyota, Nissan
├─ Colour       -> White, Blue, Red, Green, Black
├─ Doors        -> 3, 4, 5
└─ Odometer     -> already numeric

ColumnTransformer Plan
│
├─ OneHotEncoder -> Make, Colour, Doors
└─ Passthrough   -> Odometer (KM)

Result: transformed_X
│
└─ all columns are now numeric
```

---

### Read This Part Like English

```python
("one_hot", one_hot, categorical_features)
```

| Piece | Meaning |
|---|---|
| `"one_hot"` | Just the name of this step |
| `one_hot` | The actual `OneHotEncoder()` tool |
| `categorical_features` | The columns to encode |

So this means:

> Create a step called `one_hot`, use the encoder tool, and apply it to `Make`, `Colour`, and `Doors`.

And this part:

```python
remainder="passthrough"
```

means:

> Keep any columns not listed above exactly as they are.


---

### Tiny Before and After Example

**Before**

| Make | Colour | Doors | Odometer (KM) |
|---|---|---:|---:|
| Honda | White | 4 | 35431 |

**After encoding**

| Make_BMW | Make_Honda | Make_Nissan | Make_Toyota | Colour_White | Doors_4 | Odometer (KM) |
|---:|---:|---:|---:|---:|---:|---:|
| 0 | 1 | 0 | 0 | 1 | 1 | 35431 |

The exact real output has more columns, but the idea is always the same:

- categories become `0` and `1` columns
- numeric columns can stay as they are

---

### Why Does It Become 13 Columns?

Because **one-hot encoding creates one new column for each category**.

From this dataset:

- `Make` has `4` categories: BMW, Honda, Nissan, Toyota
- `Colour` has `5` categories: Black, Blue, Green, Red, White
- `Doors` has `3` categories: 3, 4, 5
- `Odometer (KM)` is passed through as `1` numeric column

So the total is:

$$4 + 5 + 3 + 1 = 13$$

That is why `transformed_X` ends up with **13 columns**.

### One Real 13-Column Example

If one original row is:

| Make | Colour | Doors | Odometer (KM) |
|---|---|---:|---:|
| Honda | White | 4 | 35431 |

then the expanded 13-column version can look like this:

| Make_BMW | Make_Honda | Make_Nissan | Make_Toyota | Colour_Black | Colour_Blue | Colour_Green | Colour_Red | Colour_White | Doors_3 | Doors_4 | Doors_5 | Odometer (KM) |
|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|
| 0 | 1 | 0 | 0 | 0 | 0 | 0 | 0 | 1 | 0 | 1 | 0 | 35431 |

That table is the clearest way to see where the **13 columns** come from:

- `4` columns for `Make`
- `5` columns for `Colour`
- `3` columns for `Doors`
- `1` column for `Odometer (KM)`

---

### Most Important Takeaway

`transformer = ColumnTransformer(...)` means: **Here is the plan.**

`transformed_X = transformer.fit_transform(X)` means: **Now apply the plan to the data.**

- `X` stays as the original table
- `transformed_X` becomes the new numeric version the model can use

### Feature Scaling, for Now

**What is feature scaling?**

Feature scaling means resizing numeric features so they live on a similar scale.
Here, `Odometer (KM)` has very large values, so we scale it with `StandardScaler()`.

**Why use it here?**

- to preview how scaling is done in a real preprocessing workflow
- to stop large numeric values from dominating in models that care about scale
- to practice applying different preprocessing steps to different columns

**Why only `Odometer (KM)`?**

- `Make`, `Colour`, and `Doors` are handled with one-hot encoding
- `Odometer (KM)` is the main numeric feature, so it is the one being scaled

> In this section, scaling is mainly an **experiment / preview**. We will learn when it is truly needed in upcoming sections.

**Quick note:**
Tree-based models like `RandomForestRegressor` usually do **not** need feature scaling much, but trying it here helps you understand the workflow.

##### **🌳 Decision Tree**

---

## What is it?

A decision tree makes predictions by asking a **series of yes/no questions**,
one after another, until it reaches a final answer.

---

## Real Life First

```
Should I carry an umbrella?

Is it cloudy?
├── NO  → Don't carry umbrella 🌤️
└── YES → Is rain forecasted?
               ├── NO  → Don't carry umbrella 🌥️
               └── YES → Carry umbrella ☂️
```

---

## In ML (Heart Disease Example)

```
Does this patient have heart disease?

Is age > 55?
├── NO  → Mostly healthy 🟢
└── YES → Is chest pain type = 3?
               ├── NO  → Mostly healthy 🟢
               └── YES → Likely sick 🔴
```

The tree finds the **best question** at each step by trying every
possible split and measuring which one separates the data most cleanly.

---

## The Vocabulary

```
             [Is age > 55?]          ← ROOT NODE   (first question)
             /             \
          [NO]             [YES]
           |                 |
       [Healthy]   [Is chest pain = 3?]  ← INTERNAL NODE (middle question)
                    /              \
                 [NO]             [YES]
                   |                |
              [Healthy] 🟢      [Sick] 🔴   ← LEAF NODES (final answer)
```

| Term          | Meaning                                      |
|---------------|----------------------------------------------|
| Root Node     | The very first question (best split overall) |
| Internal Node | Any question in the middle                   |
| Leaf Node     | Final answer — no more splitting             |
| Depth         | How many levels of questions the tree has    |

---

## How Does It Pick Questions?

### Purity — the core idea

Think of bags of red and blue balls:

```
Bag A:  50 red, 50 blue  →  Maximum confusion   (IMPURE)
Bag B:  90 red, 10 blue  →  Mostly guessable    (LESS PURE)
Bag C: 100 red,  0 blue  →  Perfect, always know (PURE)
```

**The tree wants to create bags that look like Bag C.**

At every step it asks: *"Which question splits my data into the purest bags?"*

### The measurement — Gini Impurity

| Gini = 0   | Perfectly pure  (all one class)     |
|------------|-------------------------------------|
| Gini = 0.5 | Maximally impure (50/50 split)      |

The tree tries **every possible question on every feature**,
calculates Gini for each split, and picks the one that
**lowers Gini the most**.


---

## For Regression (Predicting Numbers)

Same structure. Same splitting logic.
But the leaf doesn't say "sick/healthy" —
it says the **average value** of all training samples that ended up there.

---

## The Big Problem — Overfitting

A single tree, if grown too deep, **memorizes** the training data.

```python
clf.score(X_train, y_train)  # → 1.0   (100% — memorized)
clf.score(X_test, y_test)    # → 0.85  (drops on new data)
```

The gap between training and test score = **overfitting**.

Like a student who memorizes past exam papers word for word —
then fails when the actual exam has slightly different questions.

> ✅ This is exactly the problem **Random Forest** was invented to solve.

---

## One Line Summary

> A decision tree asks the most useful yes/no question at each step,
> measured by how purely it separates the data,
> until the answer is clear enough to stop.

## 🌲 Random Forest — Intuition

---

### 🌳 Why One Tree Is Not Enough

A single decision tree, grown deep enough, **memorizes** the training data.
It fits perfectly on what it has seen — but struggles on new data.
This is called **overfitting**.

```python
clf.score(X_train, y_train)  # → 1.0   (memorized)
clf.score(X_test, y_test)    # → 0.85  (drops on new data)
```

> **One tree is too opinionated.**
> It saw one version of the data and built one rigid structure.

---

### 🗳️ The Solution — Ask Many Trees

> Don't rely on one tree. Build hundreds of trees and let them **vote**.

This is the same logic as asking many friends for advice.
No single friend is always right — but the **majority opinion** tends to be far more accurate.

---

### 🎲 Where the "Random" Comes From

If all trees trained on the same data → they'd all build the same tree.
Voting 100 identical trees would be pointless.

So Random Forest introduces **two deliberate sources of randomness.**

---

**Randomness 1 — Different Data for Each Tree (Bootstrapping)**

Each tree gets a **random sample** of the training data, with replacement.
The same row can appear multiple times. Some rows won't appear at all.

```
Full data:    [P1, P2, P3, P4, P5, P6, P7, P8, P9, P10]

Tree 1 sees:  [P1, P1, P3, P5, P5, P7, P8, P10, P10, P2]
Tree 2 sees:  [P2, P4, P4, P6, P7, P7, P9,  P1,  P3,  P8]
Tree 3 sees:  [P3, P5, P6, P6, P8, P9, P10, P2,  P4,  P7]
```

Every tree trains on a slightly different version of reality.
So every tree builds a slightly different structure.

#### **Bootstarpping**
- Bootstrapping means: 
  - Make a new dataset by randomly picking rows from your original dataset with replacement.

- “With replacement” means:
  - After you pick a row, you put it back before the next pick.
    - the same row can appear multiple times,
    - some rows might not appear at all.
---

**Randomness 2 — Different Features at Each Split**

At every split, the algorithm doesn't consider **all** features.
It picks a **random subset** — controlled by `max_features='sqrt'`.

```
All features available → [age, cp, chol, thalach, slope, ca, thal ...]
Random subset chosen   → [age, chol, thal]
Best question picked from ONLY those 3
```

No single feature can dominate every tree.

---

### 🗳️ How the Voting Works

**Classification** → majority vote

```
Tree 1  →  Heart Disease ✓
Tree 2  →  Heart Disease ✓
Tree 3  →  No Disease    ✗
Tree 4  →  Heart Disease ✓
Tree 5  →  Heart Disease ✓

Final   →  Heart Disease  (4 out of 5)
```

**Regression** → average of all predictions

```
Tree 1  →  $320,000
Tree 2  →  $305,000
Tree 3  →  $340,000

Final   →  $321,667
```

---

### ✅ Why Overfitting Gets Fixed

Each tree still overfits — but to a **different slice of the data**.

When they vote:
- The **noise cancels out** — each tree's noise is different
- The **signal survives** — real patterns appear in most trees

---

### 🗺️ The Full Picture

```
            Original Training Data
                     ↓
       ┌─────────────┼─────────────┐
       ↓             ↓             ↓
 Random Sample  Random Sample  Random Sample  ...
       ↓             ↓             ↓
     Tree 1        Tree 2        Tree 3        ...
 (rand features)(rand features)(rand features)
       ↓             ↓             ↓
 Prediction 1  Prediction 2  Prediction 3
          ↘         ↓         ↙
             Vote / Average
                  ↓
           Final Prediction
```

---

### ⚙️ Hyperparameters Decoded

| Hyperparameter     | What it controls                                 |
|--------------------|--------------------------------------------------|
| `n_estimators=100` | How many trees to build                          |
| `max_depth`        | How deep each individual tree can grow           |
| `max_features`     | How many features considered at each split       |
| `bootstrap=True`   | Whether to use random sampling w/ replacement    |
| `random_state`     | Fixes randomness so results are reproducible     |

---

### 💡 One Line Summary

> Random Forest builds many decision trees, each trained on slightly different data
> and features, then combines their answers — so individual errors cancel out
> and the final prediction is far more reliable than any single tree.

- **Classifier** → predicts categories → `RandomForestClassifier`
- **Regressor**  → predicts numbers   → `RandomForestRegressor`

##### What is R² (Coefficient of Determination)?
> Returned by `model.score()` for regression models

---

### 1. The Main Idea

R² answers one question:

> How much better is my model than a very dumb guess?

That dumb guess is:

> Always predict the average of the target values.

---

### 2. Start With the Baseline

Imagine we are predicting house prices.

```text
Actual prices:  $300k   $350k   $400k   $450k
```

First, find the average:

```text
Average = (300 + 350 + 400 + 450) / 4
        = $375k
```

So the baseline prediction is:

```text
For every house, guess $375k
```

This is the reference point R² compares against.

---

### 3. Baseline Errors

Now check how wrong that baseline is.

```text
House 1: actual $300k, guessed $375k  -> error = 75k
House 2: actual $350k, guessed $375k  -> error = 25k
House 3: actual $400k, guessed $375k  -> error = 25k
House 4: actual $450k, guessed $375k  -> error = 75k
```

In R², we square the errors:

```text
(75)^2 + (25)^2 + (25)^2 + (75)^2
= 5625 + 625 + 625 + 5625
= 12500
```

This is the baseline total error.

---

### 4. Model Guesses

Now suppose your model predicts:

```text
House 1 -> $310k
House 2 -> $360k
House 3 -> $395k
House 4 -> $445k
```

Now compute the model's errors:

```text
House 1: actual $300k, predicted $310k  -> error = 10k
House 2: actual $350k, predicted $360k  -> error = 10k
House 3: actual $400k, predicted $395k  -> error = 5k
House 4: actual $450k, predicted $445k  -> error = 5k
```

Square them:

```text
(10)^2 + (10)^2 + (5)^2 + (5)^2
= 100 + 100 + 25 + 25
= 250
```

This is the model's total error.

---

### 5. The Formula

```text
R² = 1 - (model error / baseline error)
```

Plug in the numbers:

```text
R² = 1 - (250 / 12500)
   = 1 - 0.02
   = 0.98
```

Meaning:

> The model explains 98% of the variation in house prices.

---

### 6. How to Read R²

| R² value | Meaning |
|---|---|
| `1.0` | Perfect predictions |
| `0.8` to `0.99` | Very strong model |
| `0.5` to `0.79` | Useful, but not great |
| `0.0` | No better than guessing the average |
| Negative | Worse than guessing the average |

---

### 7. Your Ridge Model Score

Your notebook showed:

```text
R² = 0.5758
```

That means:

1. Your model is better than the baseline.
2. It explains about 57.58% of the variation in the target.
3. About 42.42% is still unexplained.

That unexplained part could come from:

- noise
- missing features
- imperfect model choice

---

### 8. R² vs Accuracy

- **For regression:**

```python

```

returns:


So both use `.score()`, but they measure different things.

#### **Predict vs Predict_Proba: Final Decision vs Model Confidence**

When working with a **classification model**, these two methods answer **different questions**:

| Method | Main question it answers | What it returns | Example output | Best use |
|---|---|---|---|---|
| `predict()` | Which class should the model choose? | Final class label | `[1, 0, 1]` | When you want the direct prediction |
| `predict_proba()` | How confident is the model about each class? | Probability for every class | `[[0.20, 0.80], [0.91, 0.09], [0.35, 0.65]]` | When you want confidence and uncertainty |

---

#### 1. `predict()` -> the model's final answer

This method gives the **chosen class label** directly.

For a heart disease example:

- `0` = No heart disease
- `1` = Heart disease

```python
clf.predict(X_test[:3])
# array([1, 0, 1])
```

That means the model's final decisions are:

- first patient -> class `1`
- second patient -> class `0`
- third patient -> class `1`

Think of `predict()` as the model saying:

> "This is my final choice."

---

#### 2. `predict_proba()` -> the model's confidence for each class

This method gives the **probability of every possible class**.

```python
clf.predict_proba(X_test[:3])
# array([
#   [0.20, 0.80],
#   [0.91, 0.09],
#   [0.35, 0.65]
# ])
```

Read each row carefully:

- `[0.20, 0.80]` means 20% chance of class `0`, 80% chance of class `1`
- `[0.91, 0.09]` means 91% chance of class `0`, 9% chance of class `1`
- `[0.35, 0.65]` means 35% chance of class `0`, 65% chance of class `1`

Think of `predict_proba()` as the model saying:

> "Here is how confident I am about each possible answer."

---

#### 3. How they are connected

In many classification models, `predict()` chooses the class with the **highest probability** from `predict_proba()`.

Example:

- `[0.20, 0.80]` -> choose class `1`
- `[0.91, 0.09]` -> choose class `0`
- `[0.35, 0.65]` -> choose class `1`

So a simple memory trick is:

- `predict_proba()` = full confidence picture
- `predict()` = final selected answer

---

#### 4. Real-world analogy

Imagine a doctor reviewing a patient's test results:

- `predict()` = "The patient has heart disease."
- `predict_proba()` = "There is an 80% chance of heart disease and a 20% chance of no heart disease."

One gives the **decision**.
The other gives the **confidence behind the decision**.

---

#### 5. Most important takeaway

Use `predict()` when you only need the final class.
Use `predict_proba()` when you also want to understand **how certain** the model is.


> Short version: `predict()` tells you **what** the model chose. `predict_proba()` tells you **how sure** it was.

*How the cross-validation working*

#### **Cross-Validation (CV)**

> **Goal:** get a *more trustworthy* estimate of how your model will do on **new/unseen data**.

> **Important:** CV does **not** magically make the model better — it makes the **evaluation** less dependent on luck.

---
### 1) Why do we need it? ("lucky" vs "unlucky")
If you do **one** train/test split, your score can change a lot depending on *which* rows ended up in the test set:
- **Lucky split**: test set happens to be easier → score looks **too high**
- **Unlucky split**: test set happens to be harder → score looks **too low**

> Analogy: tasting soup with one spoon. One scoop might be salty (unlucky) or bland (lucky). Multiple spoonfuls is fairer.

---
### 2) What is cross-validation?
**Cross-validation** = run several train/test experiments (in a structured way) and collect the scores.
Then you usually look at the **average** score (and how much the scores vary).

> Think: **multiple mini-tests** instead of one final test.

---
### 3) How does 5-fold CV work? (`cv=5`)
With `cv=5`, your dataset is split into **5 parts** (called *folds*):

> `[Fold 1] [Fold 2] [Fold 3] [Fold 4] [Fold 5]`

> You do 5 rounds: each fold becomes the test set once.

- Round 1: train on folds 2–5, test on fold 1
- Round 2: train on folds 1,3–5, test on fold 2
- Round 3: train on folds 1–2,4–5, test on fold 3
- Round 4: train on folds 1–3,5, test on fold 4
- Round 5: train on folds 1–4, test on fold 5

> Result: you get **5 scores** (one score per round).

> For classification, scikit-learn usually tries to keep class balance similar in each fold.

---
### 4) What does `cross_val_score` return?
```python
cross_val_score(clf, X, y)
```
This returns an array of scores, like:
`array([0.82, 0.79, 0.85, 0.80, 0.83])`

> The model was tested 5 times, so you got 5 scores.

Common summary:
```python
scores = cross_val_score(clf, X, y, cv=5)
scores.mean()  # average performance
scores.std()   # stability (lower = more consistent)
```

> If the scores vary a lot, it means performance depends heavily on which data slice you tested on.

---
### 5) Why these two lines look the same
```python
cross_val_score(clf, X, y)
```
vs
```python
cross_val_score(clf, X, y, cv=5)
```
In scikit-learn, the default is usually **5 folds**, so these are often effectively the same.

> Writing `cv=5` is just **more explicit** and easier to remember later.

---
### 6) CV vs a single train/test split (quick comparison)
- **One split**: faster, but can be noisy (lucky/unlucky)
- **Cross-validation**: slower, but gives a more reliable estimate

> Memory sentence: **CV is multiple mini-tests, so your score is less about luck.**

#### **📈 Area Under the ROC Curve (`AUC / ROC`)**

---

### What is the ROC Curve?

The **ROC curve** (Receiver Operating Characteristic) is a graph that shows how well a classification model separates the two classes (e.g., "Has heart disease" vs "No heart disease").

- The **x-axis** shows the **False Positive Rate (FPR)**
- The **y-axis** shows the **True Positive Rate (TPR)**

The curve is drawn by shifting a **threshold** and recording how TPR and FPR change.

---

#### 🧱 The 4 Building Blocks (Confusion Matrix)

| Prediction vs Reality | Model says **1** (Positive) | Model says **0** (Negative) |
|---|---|---|
| **Truth is 1** (Positive) | ✅ True Positive (TP) | ❌ False Negative (FN) |
| **Truth is 0** (Negative) | ❌ False Positive (FP) | ✅ True Negative (TN) |

---

### 📌 What is TPR (True Positive Rate)?

> **"Out of all the actual positives, how many did the model correctly catch?"**

$$TPR = \frac{TP}{TP + FN}$$

- Also called **Sensitivity** or **Recall**
- **High TPR** = The model catches most of the real positives
- Example: Out of 100 people who actually have heart disease, the model correctly identified 90 → TPR = 90%

---

### 📌 What is FPR (False Positive Rate)?

> **"Out of all the actual negatives, how many did the model wrongly label as positive?"**

$$FPR = \frac{FP}{FP + TN}$$

- **Low FPR** = The model doesn't raise many false alarms
- Example: Out of 100 people who do NOT have heart disease, the model wrongly flagged 10 of them → FPR = 10%

---

### 📌 What is the Threshold?

A classifier doesn't just predict "yes" or "no" — it first outputs a **probability** between 0 and 1.

The **threshold** is the cutoff point you choose to decide:

> "If the probability is **above** this number → predict **1** (Positive)"  
> "If the probability is **below** this number → predict **0** (Negative)"

| Threshold | Effect |
|---|---|
| Very **low** (e.g. 0.1) | Model says "positive" to almost everyone → High TPR, but also High FPR |
| Very **high** (e.g. 0.9) | Model is very strict → Low FPR, but also Low TPR (misses real positives) |
| **Default** = 0.5 | Balanced middle point |

The ROC curve is created by **trying every possible threshold** and plotting the resulting (FPR, TPR) point.

---

### 🏆 What is AUC?

**AUC** = **Area Under the Curve** — a single number that summarizes the whole ROC curve.

| AUC Score | Meaning |
|---|---|
| **1.0** | Perfect model |
| **0.5** | Random guessing (useless) |
| **< 0.5** | Worse than random |

> The closer AUC is to **1.0**, the better the model is at separating the two classes.

---

### 📚 **Notes**

> ROC curves and AUC metrics are evaluation metrics for **binary classification models**  
> (a model which predicts one thing or another, such as heart disease or not).

**Key takeaways:**

1. The **ROC curve** compares **TPR vs FPR** at different classification thresholds
2. A **perfect model** gets an AUC score of **1.0**
3. A model that **randomly guesses** gets an AUC of **0.5** (no better than a coin flip)

#### **ROC, AUC, FPR, TPR, Thresholds**

> Simple goal: turn model probabilities into clear decisions and measure trade-offs correctly.

### 1) What `predict_proba` gives
- For binary classification, each row is usually:
  - `[P(class 0), P(class 1)]`
- Always verify class order with `clf.classes_`.
  - If `clf.classes_ == [0, 1]`, then column 1 is probability of class 1.

### 2) What a threshold is
- Threshold = cutoff to convert probability into label.
- Rule: if `P(class 1) >= threshold`, predict `1`; else predict `0`.
- Lower threshold -> more predicted 1s.
- Higher threshold -> fewer predicted 1s.

### 3) How TP, TN, FP, FN are counted
- **TP**: predicted 1, actual 1
- **TN**: predicted 0, actual 0
- **FP**: predicted 1, actual 0
- **FN**: predicted 0, actual 1

### 4) Core formulas (most important)
$$
TPR = \frac{TP}{TP + FN}
$$
- Of all actual positives, how many we correctly caught.

$$
FPR = \frac{FP}{FP + TN}
$$
- Of all actual negatives, how many we wrongly flagged as positive.

Memory trick:
- TPR denominator = all actual positives (`TP + FN`)
- FPR denominator = all actual negatives (`FP + TN`)

### 5) `roc_curve` output order and meaning

```python
fpr, tpr, thresholds = roc_curve(y_test, y_probs_positive)
```
- Function returns arrays in this order: **fpr, tpr, thresholds**.
- But each index belongs together:
  - `thresholds[i]` produced `fpr[i]` and `tpr[i]`.

So at any index `i`, think:
- cutoff used = `thresholds[i]`
- false-positive rate = `fpr[i]`
- true-positive rate = `tpr[i]`

### 6) Why first threshold is often `inf`
- `inf` means a super-strict cutoff.
- Almost nothing is predicted as class 1.
- So the first point is usually near:
  - `TPR = 0`, `FPR = 0`

### 7) ROC vs AUC
- **ROC**: the full curve of points `(FPR, TPR)` across thresholds.
- **AUC**: one summary number = area under ROC.
  - Higher AUC is better.
  - `AUC = 0.5` is like random guessing.
  - `AUC = 1.0` is perfect separation.

### 8) One-line intuition
- Lower threshold usually increases both TPR and FPR.
- Higher threshold usually decreases both TPR and FPR.

### 9) Practical mini-workflow
1. Train model.
2. Get positive probabilities: `y_probs = clf.predict_proba(X_test)[:, 1]`.
3. Get ROC arrays: `fpr, tpr, thresholds = roc_curve(y_test, y_probs)`.
4. Optionally compute single score: `roc_auc_score(y_test, y_probs)`.

---
If you forget later, come back to this page and only remember this first:
**same index ties threshold, FPR, and TPR together.**

````python
# Create a function for plotting ROC curves

import matplotlib.pyplot as plt

def plot_roc_curve(fpr, tpr):
    """
    Plots a ROC curve given the false positive rate(fpr)
    and true positive rate(tpr)
    """
    
    # Plot the ROC curve``
    plt.plot(fpr, tpr, color="orange", label="ROC curve")
    
    # Plot line with no predictive power (baseline)
    plt.plot([0, 1], [0, 1], color="darkblue", linestyle="--", label="Guessing")
    
    # Customize the plot
    plt.xlabel("False positive rate (FPR)")
    plt.ylabel("True positive rate (TPR)")
    plt.title("Receiver Operating Characteristic (ROC) Curve")
    plt.legend()
    plt.show()    
    
plot_roc_curve(fpr, tpr)
````

#### **Confusion Matrix**


> A confusion matrix is a quick way to compare the labels a model predicts and the actual labels it was supposed to predict.

**It compares:**

1. What the model predicted
2. What the real answer was


**So it tells you exactly where model is right and wrong.**

![Confusion Matrix](../assets/images/Confusion_matrix.png)

**For a Yes/No problem:**

1. **True Positive (TP)**
    
    Model says Yes, and real answer is Yes.
    > Good.

2. **True Negative (TN)**
    
    Model says No, and real answer is No.
    > Good.

3. **False Positive (FP)**
    
    Model says Yes, but real answer is No.
    > False alarm.

4. **False Negative (FN)**
    
    Model says No, but real answer is Yes.
    >Missed case.

**How to Read a Confusion Matrix (Diagram)**

```text
                    Predicted
                 0 (No)      1 (Yes)
Actual 0 (No)   [ TN ]       [ FP ]
Actual 1 (Yes)  [ FN ]       [ TP ]
```

```text
Index map (array form):
cm[0,0] = TN   cm[0,1] = FP
cm[1,0] = FN   cm[1,1] = TP
```

- Top-left = True Negative (correct No)
- Top-right = False Positive (false alarm)
- Bottom-left = False Negative (missed Yes)
- Bottom-right = True Positive (correct Yes)

### **Classification Report**

A **classification report** gives you a detailed breakdown of how well your model performs on each class, with multiple metrics beyond just accuracy.

---


#### What metrics does it show?

A classification report displays four main metrics for each class:

| Metric | What it means | Formula |
|---|---|---|
| **Precision** | Of all the times the model predicted this class, how many were correct? | `TP / (TP + FP)` |
| **Recall (TPR)** | Of all the actual instances of this class, how many did the model catch? | `TP / (TP + FN)` |
| **F1-Score** | A balanced average of Precision and Recall | `2 * (Precision * Recall) / (Precision + Recall)` |
| **Support** | How many actual instances of this class exist in the test set | Count |

---

## 1️⃣ **Precision** — Avoiding False Alarms

> "When the model says YES, how often is it right?"

**Example: Heart Disease Prediction**

```
Model predicts "Has Heart Disease" 100 times
Actually has disease: 80 times
Doesn't have disease: 20 times (false alarms)

Precision = 80 / 100 = 0.80 (80%)
```

**Use precision when:** False positives are costly
- Spam filter (don't want to block real emails)
- Medical diagnosis (don't want to scare healthy people)

---

## 2️⃣ **Recall (Also called TPR)** — Not Missing Cases

> "When the actual answer is YES, how often does the model catch it?"

**Example: Heart Disease Prediction (same scenario)**

```
Actually has disease: 100 people total
Model correctly identified: 80 people
Model missed: 20 people

Recall = TPR = 80 / 100 = 0.80 (80%)
```

**Use recall when:** False negatives are costly
- Cancer screening (missing a case is dangerous)
- Security (missing a threat is bad)

---

## 3️⃣ **The Precision vs Recall Trade-off**

You usually **can't have both at 100%**. There's a trade-off:

| Scenario | Precision High | Recall High |
|---|---|---|
| Model behavior | Very strict, only predicts YES when confident | Generous, predicts YES often |
| Result | Few false alarms, but misses some cases | Catches most cases, but raises many false alarms |
| Threshold effect | High threshold | Low threshold |
| Use case | Spam filter | Cancer screening |

```
         High Threshold (Strict)
              ↓
    Precision ↑    Recall ↓
    
         Low Threshold (Generous)
              ↓
    Precision ↓    Recall ↑
```

---

## 4️⃣ **F1-Score** — The Balanced Middle Ground

> "Give me one number that balances both precision and recall."

The **F1-score** is the **harmonic mean** of precision and recall (0 to 1).

- **F1 = 1.0** → Perfect precision and recall
- **F1 = 0.5** → Balanced but mediocre
- **F1 = 0.0** → Terrible on both metrics

**Use F1-score when:** You care equally about false positives AND false negatives.

---

## 5️⃣ **Support** — The Class Count

> "How many actual instances of this class are in my test set?"

**Example:**

```
Class 0 (No Disease): 32 samples
Class 1 (Disease): 29 samples
Total support: 61 samples
```

Support tells you if your dataset is **balanced** or **imbalanced**:
- Balanced: similar support for each class ✅
- Imbalanced: very different support (e.g., 90 vs 10) ⚠️

---

## 6️⃣ **How to Read a Classification Report**

**Real example output:**

```
              precision    recall  f1-score   support

           0       0.86      0.89      0.88        32
           1       0.87      0.83      0.85        29

    accuracy                           0.86        61
   macro avg       0.87      0.86      0.86        61
weighted avg       0.87      0.86      0.86        61
```

**Reading this:**

| Line | Meaning |
|---|---|
| `0` | Metrics for class 0 (No disease) |
| `1` | Metrics for class 1 (Disease) |
| `accuracy` | Overall accuracy across all classes |
| `macro avg` | Simple average of metrics (treats classes equally) |
| `weighted avg` | Weighted average (accounts for support in each class) |

---

## 7️⃣ **Macro vs Weighted Average**

Both average the metrics across classes, but differently:

```
Class 0: Precision = 0.90, Support = 100
Class 1: Precision = 0.70, Support = 10

Macro avg     = (0.90 + 0.70) / 2 = 0.80
Weighted avg  = (0.90*100 + 0.70*10) / (100+10) = 0.87
```

- **Macro average**: treats each class equally (ignores imbalance)
- **Weighted average**: favors the class with more samples

**Use weighted average** if your dataset is imbalanced.

#### **R^2**

##### **What is R² (Coefficient of Determination)?**
> Returned by `model.score()` for regression models

---

### 1. The Main Idea

R² answers one question:

> How much better is my model than a very dumb guess?

That dumb guess is:

> Always predict the average of the target values.

---

### 2. Start With the Baseline

Imagine we are predicting house prices.

```text
Actual prices:  $300k   $350k   $400k   $450k
```

First, find the average:

```text
Average = (300 + 350 + 400 + 450) / 4
        = $375k
```

So the baseline prediction is:

```text
For every house, guess $375k
```

This is the reference point R² compares against.

---

### 3. Baseline Errors

Now check how wrong that baseline is.

```text
House 1: actual $300k, guessed $375k  -> error = 75k
House 2: actual $350k, guessed $375k  -> error = 25k
House 3: actual $400k, guessed $375k  -> error = 25k
House 4: actual $450k, guessed $375k  -> error = 75k
```

In R², we square the errors:

```text
(75)^2 + (25)^2 + (25)^2 + (75)^2
= 5625 + 625 + 625 + 5625
= 12500
```

This is the baseline total error.

---

### 4. Model Guesses

Now suppose your model predicts:

```text
House 1 -> $310k
House 2 -> $360k
House 3 -> $395k
House 4 -> $445k
```

Now compute the model's errors:

```text
House 1: actual $300k, predicted $310k  -> error = 10k
House 2: actual $350k, predicted $360k  -> error = 10k
House 3: actual $400k, predicted $395k  -> error = 5k
House 4: actual $450k, predicted $445k  -> error = 5k
```

Square them:

```text
(10)^2 + (10)^2 + (5)^2 + (5)^2
= 100 + 100 + 25 + 25
= 250
```

This is the model's total error.

---

### 5. The Formula

```text
R² = 1 - (model error / baseline error)
```

Plug in the numbers:

```text
R² = 1 - (250 / 12500)
   = 1 - 0.02
   = 0.98
```

Meaning:

> The model explains 98% of the variation in house prices.

---

### 6. How to Read R²

| R² value | Meaning |
|---|---|
| `1.0` | Perfect predictions |
| `0.8` to `0.99` | Very strong model |
| `0.5` to `0.79` | Useful, but not great |
| `0.0` | No better than guessing the average |
| Negative | Worse than guessing the average |

---

### 7. Your Ridge Model Score

Your notebook showed:

```text

```

That means:

1. Your model is better than the baseline.
2. It explains about 57.58% of the variation in the target.
3. About 42.42% is still unexplained.

That unexplained part could come from:

- noise
- missing features
- imperfect model choice
- limited data

#### **Mean Absolute Error (MAE)**

**Big idea:** MAE tells you the model’s *average miss distance* (how far off it is), ignoring whether the model was too high or too low.

**Visual (what MAE does):**

> `Actual (y)` → compare with → `Predicted (ŷ)` → get → `Error (y - ŷ)` → take → `|Error|` → average them

**How to compute MAE (3 steps):**
1. For each data point, compute the error: `actual - predicted`
2. Make it positive: `abs(actual - predicted)`
3. Average all those absolute errors

<details>
<summary><strong>Formula (optional)</strong></summary>

> $$\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}\left|y_i - \hat{y}_i\right|$$
</details>

**Quick notes:**
- MAE is the average of the absolute differences between predictions and actual values.
- Lower MAE is better.
- MAE is in the **same unit** as the target (dollars, kg, etc.).
- It gives you an idea of how wrong your models predictions are.
  

**Quick example**

- Actual: `[100, 200, 300]`
- Predicted: `[110, 190, 260]`

- Errors (`actual - predicted`): `[-10, 10, 40]`
- Absolute errors: `[10, 10, 40]`

`MAE = (10 + 10 + 40) / 3 = 20`

✅ **Interpretation:** The model is off by about **20 units** on average.

#### **Mean Sqaured Error (MSE)**

MSE measures the **average of squared prediction errors**. It's harsher than MAE because it **penalizes big mistakes much more**.

$$\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

> MSE is the mean of the square of the errors between actual and predicted values.

##### **1) Why Square the Errors?**

Squaring errors serves two purposes:
1. **Removes negative/positive signs** (just like MAE's absolute value)
2. **Amplifies big mistakes** (the bigger the error, the worse it looks)

**Real example:**
| Error | Contribution to MAE | Contribution to MSE |
|---|---|---|
| 2 | 2 | 4 |
| 10 | 10 | 100 |
| 50 | 50 | 2500 |

See how MSE **explodes** with big errors, while MAE grows linearly? That's the power of squaring.

---

##### **2) Working Example (Step-by-Step)**

**Predictions:**
- Actual: `[10, 20, 30]`
- Predicted: `[12, 19, 25]`

**Step 1: Compute errors**
- Error 1: 10 - 12 = **-2**
- Error 2: 20 - 19 = **+1**
- Error 3: 30 - 25 = **+5**

**Step 2: Square each error**
- (-2)² = **4**
- (+1)² = **1**
- (+5)² = **25**

**Step 3: Average**
- MSE = (4 + 1 + 25) / 3 = **10**

---

##### **3) Formula and Units (Important!)**

$$\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

**⚠️ Units gotcha:** If your target is in **dollars**, MSE is in **dollars²** (awkward!). That's why people often use **RMSE** = √MSE to get back to original units.

---

##### **4) MSE vs MAE — Quick Comparison**

| Aspect | MAE | MSE |
|---|---|---|
| **Formula** | \|y - ŷ\| averaged | (y - ŷ)² averaged |
| **Big errors** | Counted normally | Counted much MORE |
| **Units** | Same as target ✓ | Target² (awkward) ✗ |
| **Interpretation** | "Average miss distance" | "Average squared miss" |
| **Sensitivity** | Normal | Ultra-sensitive to outliers |

---

##### **6) When to Use MSE**

✅ **Use MSE when:**
- You want to **heavily penalize big mistakes**
- One huge error is much worse than many small ones
- Examples: house price prediction (off by $200k = disaster), stock prices, medical dosing

❌ **Don't use MSE when:**
- Big outliers exist and you want to ignore them
- You need a human-friendly metric in original units

---

### **7) When to Use MAE**

✅ **Use MAE when:**
- You want **simple, interpretable results** (average error in original units)
- Errors are roughly equally important (small or large)
- You don't care about outliers as much

❌ **Don't use MAE when:**
- Big mistakes need to be heavily punished
- You need to match what others are using (MSE is more common in research)

---

### **8) Memory Tricks**

> **MAE** = Friendly, tells you "average miss distance" → use for interpretation  
> **MSE** = Strict teacher, hates big mistakes → use when outliers are bad  
> **RMSE** = Same as MSE but back in original units (√MSE)

---

### **9) Real Scenario**

Imagine predicting delivery times:
- **With MAE:** Errors of 2, 3, 5 hours → average 3.3 hours off per delivery ✓ (easy to understand)
- **With MSE:** Same errors → MSE = 38, which doesn't tell you much ✗

But if one crazy prediction is 100 hours off:
- **With MAE:** Still "reasonable" (lowers average by misses a bit)
- **With MSE:** Explodes to huge number, forcing you to **fix that outlier** ✓ (catches problems)

### **Now you might be thinking, which regression evaluation metric should you use?**

- `R^2` is similar to accuracy. It gives you a quick indication of how well your model might be doing. Generally, the closer your `R^2` value is to 1.0, the better the model. But it doesn't really tell exactly how wrong your model is in terms of how far off each prediction is.
- `MAE` gives a better indication of how far off each of your model's predictions are on average.
- As for `MAE` or `MSE`, because of the way `MSE` is calculated, squaring the differences between predicted values and actual values, it amplifies larger differences. Let's say we're predicting the value of houses (which we are).
- **Pay more attention to `MAE`:** When being $10,000 off is twice as bad as being $5,000 off.
- **Pay more attention to `MSE`:** When being $10,000 off is more than twice as bad as being $5,000 off.

````python
# Creating a Evaluation Function since we need to evaluate multiple times
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluate_preds(y_true, y_preds):
    """
    Performs evaluation comparison on y_true labels vs y_pred labels.
    Use this for a Classification model
    """
    
    accuracy = accuracy_score(y_true, y_preds)
    precison = precision_score(y_true, y_preds)
    recall = recall_score(y_true, y_preds)
    f1 = f1_score(y_true, y_preds)
    
    metric_dict = {"accuracy" : round(accuracy, 2),
                   "precision" : round(precison, 2),
                   "recall" : round(recall, 2),
                   "f1" : round(f1, 2)}
    
    print(f"Accuracy: {accuracy * 100:.2f}%")
    print(f"Precision: {precison:.2f}")
    print(f"Recall: {recall:.2f}")
    print(f"F1 Score: {f1:.2f}")
    
    return metric_dict
````

````python
from sklearn.ensemble import RandomForestClassifier

np.random.seed(42)

# Shuffle the data
heart_disease_shuffled = heart_disease.sample(frac=1) # It shuffles all the rows in the heart_disease DataFrame randomly, but keeps all the data (no rows are dropped)(100% data =  frac = 1).

# Spilt into X and y
X = heart_disease_shuffled.drop('target', axis=1)
y = heart_disease_shuffled['target']

# Split the data into train, validation, & test set
train_split = round(0.7 * len(heart_disease_shuffled)) # 70% of data
valid_split = round(train_split + 0.15 * len(heart_disease_shuffled)) # 15% of data

X_train, y_train = X[:train_split], y[:train_split]
X_valid, y_valid = X[train_split: valid_split], y[train_split : valid_split]
X_test, y_test = X[valid_split:], y[valid_split:]

# Create and train a model
clf = RandomForestClassifier()
clf.fit(X_train, y_train)

# Make baseline predictions
y_preds = clf.predict(X_valid)

# Evaluate the classifier on validation set
baseline_metrics = evaluate_preds(y_valid, y_preds)
````

- How deep each tree can grow.

```python
RandomForestClassifier(max_depth=10)
```

- `None` → tree grows until all leaves are pure (can overfit)
- Setting a number → caps the depth, forces simpler trees
- **Think of it as:** how many questions the tree is allowed to ask

3. **`max_features`(default: "`sqrt`")**

How many features each tree considers when looking for the best split.

```python
RandomForestClassifier(max_features="sqrt")
```

1. `"sqrt"` --> √(total features) — default, works well for classification
2. `"log2"` --> log₂(total features) — even fewer features
3. `0.5` --> 50% of features
4. `None` --> All features — removes feature randomness


- This is the most impactful randomness hyperparameter
- Smaller value = more random, more diverse trees
- **Rule of thumb:** Keep "sqrt" for classification

---

4. **`min_samples_split` (default: 2)**

Minimum number of samples a node must have to be allowed to split further.

```python
RandomForestClassifier(min_samples_split=5)
```

- Higher value → tree stops splitting sooner → simpler tree
- Helps prevent overfitting on tiny groups of data
- **Think of it as:** "don't split a group unless it has at least N people in it"

5. **`min_samples_leaf`(default: 1)**
   
Minimum number of samples required to be at a leaf (end node).

```python
RandomForestClassifier(min_samples_leaf=2)
```

- Similar to above but controls the end of the tree, not the middle
- Higher value → smoother predictions, less overfitting
- **Think of it as:** "every final answer must represent at least N people"

### **5.2 Hyperparameter Tuning with `RandomizedSearchCV`**

When you create a model, sklearn uses default values for all its hyperparameters. These defaults are okay but not necessarily the best for your specific dataset. **Hyperparameter tuning** is the process of searching for the best combination of settings for your data.

Scikit-Learn's [sklearn.model_selection.RandomizedSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.RandomizedSearchCV.html) does this by **randomly sampling** combinations from a grid of options, cross-validating each one, and returning the best.

> **Key mental model:** `RandomizedSearchCV` = _"Try N random combinations, cross-validate each one, return the best."_

---

#### The Search Grid

You define the options to search over as a dictionary:

```python
param_distributions = {
    "hyperparameter_name": [values_to_randomly_try]
}
```

- `"hyperparameter_name"` — a hyperparameter the model accepts
- `[values_to_randomly_try]` — list of values to sample from for that hyperparameter

---



#### How It Works Internally

`RandomizedSearchCV` runs in two layers:

**Layer 1 — Random sampling (`n_iter`)**  
Randomly picks `n_iter` combinations from all possible ones in the grid.

**Layer 2 — Cross-validation (`cv`)**  
For each combination, it runs `cv`-fold cross-validation on your training data:

- Splits `X_train` into `cv` equal folds
- Rotates through them — each fold takes a turn as the validation set
- Averages the scores → one final score for that combination

Total models trained = `n_iter × cv`

---

#### The Three-Way Split (Correct Workflow)

You should always `train_test_split` **before** running `RandomizedSearchCV`:

```
Full Dataset
│
└── train_test_split()
        │
        ├── X_train ──→ goes into RandomizedSearchCV
        │                    └── cv folds create internal validation sets
        │
        └── X_test ───→ locked away — used only once at the very end
```

| Set            | Created by                  | Purpose                                               |
| -------------- | --------------------------- | ----------------------------------------------------- |
| **Train**      | `train_test_split`          | RandomizedSearchCV trains models on this              |
| **Validation** | Cross-validation (internal) | Scores each combo to find the best                    |
| **Test**       | `train_test_split`          | Final honest evaluation — never touched during search |

> `X_test` is never seen during the search. It only exists to give you one honest final score after all tuning is done.

---

#### After the Search

```python
rs_clf.best_params_      # the winning combination
rs_clf.best_estimator_   # the best model, ready to use
rs_clf.score(X_test, y_test)  # evaluate on held-out test set
```


````python
from sklearn.model_selection import RandomizedSearchCV

grid = {"n_estimators" : [10, 100, 200, 500, 1000, 1200],
        "max_depth" : [None, 5, 10, 20, 30],
        "max_features" : [None, "log2", "sqrt"],
        "min_samples_split" : [2, 4, 6],
        "min_samples_leaf" : [1, 2, 4]}

# This grid has 6 × 5 × 3 × 3 × 3 = 810 possible combinations
# trying 10 of them (n_iter=10)

np.random.seed(42)

# Split into X and y
X = heart_disease_shuffled.drop('target', axis=1)
y = heart_disease_shuffled['target']

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Instatiate RandomForestClassifier
clf = RandomForestClassifier(n_jobs=-1)

# Setup RandomizedSearchCV
rs_clf = RandomizedSearchCV(estimator=clf, # the model to tune
                            param_distributions=grid, # the knobs and their options
                            n_iter=10, # number of models to try
                            cv=5,      # cross-validation folds per combo
                            verbose=2) # print progress while running

# Fit the RandomizedSearchCV version of clf
rs_clf.fit(X_train, y_train)
````

- **`estimator`** — Which model to tune. Here it's your `RandomForestClassifier`.

- **`param_distributions`** — The "menu" of options. It picks randomly from each list.

- **`n_iter=10`** — Try only 10 random combos out of 810. Fast but less thorough. Raise this if you have time.

- **`cv=5`** — For *each* combo it tries, it runs 5-fold cross-validation and averages the score. So it trains **10 × 5 = 50 models** total.

- **`verbose=2`** — Prints each model's progress so you can see it working.

- **`n_jobs`** controls how many CPU cores sklearn uses to train trees in parallel.
```python
RandomForestClassifier(n_jobs=1)   # 1 core — slow (default)
RandomForestClassifier(n_jobs=-1)  # all available cores — recommended
```

Since each tree in a Random Forest is independent, they can be built simultaneously. Set `n_jobs=-1` in practice, especially when running `RandomizedSearchCV` where many models are being trained.

````python
rs_clf.best_params_
````

- When `RandomizedSearchCV` goes through `n_iter` combinations of hyperparameter search space, it stores the best ones in the attribute `best_params_`.
- Now when we call `predict()` on `rs_clf` (our RandomizedSearchCV version of our classifier), it'll use the best hyperparameters it found.

````python
# Make predictions with the best hyperparameters
rs_y_preds = rs_clf.predict(X_test)

# Evaluate the predictions
rs_metrics = evaluate_preds(y_test, rs_y_preds)
````

So there is not garantee that the hyperparameter that we get are the best one sometimes it can perform worse or sometimes way better than manual way of evaluating the model

### **5.3 Hyperparameter Tuning with `GridSearchCV`**

Scikit-Learn's [sklearn.model_selection.GridSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html) works exactly like `RandomizedSearchCV` — except instead of randomly sampling combinations, it tries **every single combination** in the grid.

> **Key mental model:** `GridSearchCV` = *"Try ALL combinations, cross-validate each one, return the best."*

- GridSearchCV also refers to a dictionary of parameter distributions as a parameter grid (via the parameter `param_grid`).
---

#### **RandomizedSearchCV vs GridSearchCV**
```python
grid = {
    "n_estimators": [100, 500],
    "max_depth":    [5, 10, None]
}
# 2 × 3 = 6 total combinations
```

| | `RandomizedSearchCV` | `GridSearchCV` |
|---|---|---|
| **Combinations tried** | N random ones (`n_iter`) | All of them |
| **Speed** | Fast | Slow |
| **Best for** | Large grids | Small grids |
| **Misses anything?** | Possibly | Never |

---

````python
from sklearn.model_selection import GridSearchCV, train_test_split

np.random.seed(42)

# Split into X and y
X = heart_disease_shuffled.drop('target', axis=1)
y = heart_disease_shuffled['target']

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Instatiate RandomForestClassifier
clf = RandomForestClassifier(n_jobs=-1)

# Setup RandomizedSearchCV
gs_clf = GridSearchCV(estimator=clf, # the model to tune
                      param_grid=grid_2, # the knobs and their options
                      cv=5,      # cross-validation folds per combo
                      verbose=2) # print progress while running

# Fit the RandomizedSearchCV version of clf
gs_clf.fit(X_train, y_train)
````

#### **Common Workflow in Practice**

Use both together for efficient tuning:

**Step 1 — `RandomizedSearchCV` first**  
Run a random search over your full large grid. Look at the results, notice which hyperparameter values are showing up in the high-scoring combos.
```python
# Large grid — 810 combinations
grid = {
    "n_estimators":      [10, 100, 200, 500, 1000, 1200],
    "max_depth":         [None, 5, 10, 20, 30],
    "max_features":      [None, "log2", "sqrt"],
    "min_samples_split": [2, 4, 6],
    "min_samples_leaf":  [1, 2, 4]
}
# RandomizedSearch tries 10 random combos → notices best scores cluster around
# n_estimators: 200-500 and max_depth: 10-20
```

**Step 2 — `GridSearchCV` second**  
Build a smaller, tighter grid from that observation and let GridSearch test every combination in it exhaustively.
```python
# Refined grid — only 12 combinations now
grid_refined = {
    "n_estimators": [200, 300, 400, 500],
    "max_depth":    [10, 15, 20]
}
# GridSearch tests all 12 → finds the exact best combo
```

This saves a huge amount of compute:
```
GridSearch on full grid:    810 combos × cv=5 = 4050 models
GridSearch on refined grid:  12 combos × cv=5 =   60 models
```
Nearly the same result —> at a fraction of the cost.
